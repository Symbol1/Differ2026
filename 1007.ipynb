## Chapter 3: Modeling with First-Order Equations

Modeling starts with a statement about rates.

1. Define the unknown quantity and its units.
2. Translate the rate statement into a differential equation.
3. Translate measurements into conditions.
4. Solve the equation and determine any unknown parameters.
5. Interpret the answer in the original units.

## 3.1 Linear Models

### Example 1: bacterial growth

A culture initially contains $P_0 > 0$ bacteria.  
After one hour it contains $3P_0/2$ bacteria.  
Assume that the growth rate is proportional to the population.

With $t$ measured in hours,
$$ A' = kA, \qquad A(0) = P_0, \qquad A(1) = \frac32 P_0. $$
Find the time at which $A(t) = 3P_0$.

Separation, or the linear-equation method, gives
$$ A(t) = Ce^{kt}. $$
The first measurement gives $C = P_0$; the second gives
$$ e^k = \frac32, \qquad k = \ln\frac32\ \mathrm{h}^{-1}. $$
Thus
$$ \boxed{A(t) = P_0e^{t\ln(3/2)}.} $$
The equilibrium $A = 0$ is excluded by the positive initial population.

The tripling time is
$$ \boxed{t = \frac{\ln3}{\ln(3/2)} \approx 2.71\ \mathrm{h}.} $$
Why were two measurements needed?  
There are two unknown constants: the initial scale $C$ and the model parameter $k$.  
For a known $k$, a single initial condition would determine the solution.

### Radioactive decay and half-life

Assume that the decay rate is proportional to the amount $A(t)$ remaining:
$$ A' = kA, \qquad A(0) = A_0, \qquad k < 0. $$
Then
$$ \boxed{A(t) = A_0e^{kt}.} $$
The **half-life** $T_{1/2}$ satisfies $A(T_{1/2}) = A_0/2$, so
$$ \boxed{T_{1/2} = \frac{\ln(1/2)}{k} = -\frac{\ln2}{k}.} $$

Equivalently, if the half-life is known, then
$$ \boxed{k = -\frac{\ln2}{T_{1/2}}, \qquad A(t) = A_0 2^{-t/T_{1/2}}.} $$
After $n$ half-lives, the fraction remaining is $2^{-n}$.

### Example 2: half-life of plutonium

A breeder reactor produces plutonium-239.  
After 15 years, $0.043\%$ of the initial amount has disintegrated.  
Assuming proportional decay, find its half-life.

Because $99.957\%$ remains,
$$ A(15) = 0.99957A_0. $$

Use $A(t) = A_0e^{kt}$:
$$ 0.99957A_0 = A_0e^{15k}, \qquad
k = \frac1{15}\ln(0.99957) \approx -2.8673 \times 10^{-5}\ \mathrm{yr}^{-1}. $$
The initial amount cancels; only the fraction remaining is needed.

Therefore,
$$ \boxed{T_{1/2} = \frac{\ln(1/2)}{k} \approx 24{,}180\ \mathrm{yr}.} $$
The small yearly decay rate corresponds to a long half-life.

### Carbon dating

Living organisms continually exchange carbon with the environment.  
After death, that exchange stops and the carbon-14 already present begins to decay without replacement.

Using the carbon-14 half-life $T_{1/2} = 5730$ years,
$$ \boxed{A(t) = A_0 2^{-t/5730}.} $$
Measuring the fraction $A(t)/A_0$ therefore gives an estimate of the age.

### Example 3: age of a fossil

A fossilized bone contains $0.1\%$ of its original carbon-14.  
Estimate the age of the fossil.

The remaining fraction is $0.001$, so
$$ 0.001 = 2^{-t/5730}. $$

Taking logarithms gives
$$ \boxed{t = \frac{5730\ln1000}{\ln2} \approx 57{,}100\ \mathrm{yr}.} $$
This is almost 10 half-lives, near the usual practical limit of conventional carbon-14 dating.

### Example 4: cooling a cake

A cake leaves an oven at $149^\circ\mathrm C$.  
After 3 minutes its temperature is $85^\circ\mathrm C$.  
The room stays at $21^\circ\mathrm C$.

Newton's law of cooling gives
$$ T' = k(T - 21), \qquad T(0) = 149, \qquad T(3) = 85. $$
Here $t$ is in minutes and cooling corresponds to $k < 0$.

Let $U = T - 21$. Then $U' = kU$ and $U(0) = 128$, so
$$ T(t) = 21 + 128e^{kt}. $$
Use the second measurement:
$$ 64 = 128e^{3k} \quad\Longrightarrow\quad k = -\frac{\ln2}{3}. $$
Therefore,
$$ \boxed{T(t) = 21 + 128\cdot 2^{-t/3}.} $$

To reach $22^\circ\mathrm C$,
$$ 1 = 128\cdot 2^{-t/3} = 2^{7 - t/3}, $$
so
$$ \boxed{t = 21\ \mathrm{minutes}} $$
after removal from the oven.

The linear method gives the same result using $T' - kT = -21k$  
and integrating factor $e^{-kt}$.

### Example 5: salt in a mixing tank

A well-mixed tank contains 1000 L of solution and initially 25 kg of salt.  
Brine containing 0.25 kg/L enters at 10 L/min.  
Mixture leaves at 10 L/min, so the volume stays constant.

Let $A(t)$ be the salt mass in kilograms.  
The outgoing concentration is $A(t)/1000$ kg/L.

The mass balance is
$$ \frac{dA}{dt} = \text{salt in per minute} - \text{salt out per minute}, $$
$$ A' = 10(0.25) - 10\frac{A}{1000} = 2.5 - 0.01A, \qquad A(0) = 25. $$
The equilibrium mass is $A_* = 250$ kg.

Shift by the equilibrium, or use the integrating factor $e^{0.01t}$:
$$ \boxed{A(t) = 250 - 225e^{-0.01t}\ \mathrm{kg}.} $$
The concentration is
$$ \boxed{c(t) = 0.25 - 0.225e^{-0.01t}\ \mathrm{kg/L}.} $$
It approaches the incoming concentration, $0.25$ kg/L.

### Series circuits: voltage balances

Let $i(t)$ be current, $q(t)$ charge, and $E(t)$ the source voltage.  
Then $i = dq/dt$.

| Component | Voltage drop |
|---|---|
| Resistor, resistance $R$ | $Ri$ |
| Inductor, inductance $L$ | $L\,di/dt$ |
| Capacitor, capacitance $C$ | $q/C$ |

Kirchhoff's voltage law equates their sum to the source voltage.

### LR series circuit

![Series circuit (source PDF, p. 88)](0923/de2-circuit-54.png)

An inductor and resistor in series satisfy
$$ \boxed{L\frac{di}{dt} + Ri = E(t).} $$
For constant $L > 0$, divide by $L$ and use integrating factor $e^{Rt/L}$.

For a constant source $E_0$ and $R > 0$,
$$ i(t) = \frac{E_0}{R} + \left(i(0) - \frac{E_0}{R}\right)e^{-Rt/L}. $$
The time constant is $\tau = L/R$.

#### Simulation: the current cannot jump

Switch a $12$ V source on and off every $0.1$ s across $R = 10\ \Omega$,  
and integrate
$$ L\frac{di}{dt} + Ri = E(t), \qquad i(0) = 0, $$
numerically for two inductances.  
A resistor by itself would copy the switch exactly.

```python
import numpy as np
from scipy.integrate import odeint
lr_E0, lr_R = float(12), float(10)              # 12 V source, 10 ohm resistor
lr_period, lr_on = float(0.2), float(0.1)       # on for half of each period
def lr_source(t):
    return lr_E0 if t % lr_period < lr_on else 0.0
def lr_rhs(current, t, L):
    return (lr_source(t) - lr_R*float(current[0]))/L
lr_times = np.linspace(0, 0.6, 1201)
lr_curves = [(L, color, odeint(lr_rhs, [0.0], lr_times, args=(float(L),), hmax=0.001)[:, 0])
             for L, color in [(0.1, '#0072B2'), (0.5, '#D55E00')]]
```

```python
lr_plot = line([(t, lr_source(t)/lr_R) for t in lr_times], color='silver',
               thickness=1.5, legend_label='Resistor alone')
for L, color, current in lr_curves:
    lr_plot += line(list(zip(lr_times, current)), color=color, thickness=2,
                    legend_label=rf'$L = {float(L)}$ H, $\tau = {float(L)/lr_R}$ s')
lr_plot.show(xmin=0, xmax=0.6, ymin=-0.05, ymax=1.5, figsize=[9, 4],
             axes_labels=['Time $t$ (s)', 'Current $i$ (A)'],
             frame=True, axes=False, legend_options={'loc': 'upper right'})
```

**Read the picture:** the gray steps are $E(t)/R$, what a resistor alone would do.  
Both currents are continuous: at each switching the slope starts at $E_0/L$, never a jump.  
The small inductance, $\tau = 0.01$ s, nearly fills each $0.1$ s window;  
the large one, $\tau = 0.05$ s, only reaches $86\%$ of $1.2$ A and never fully empties.  
Larger $L$ means more inertia, so the current is smoother and lags further behind.

### RC series circuit

![Series circuit (source PDF, p. 89)](0923/de2-circuit-55.png)

A resistor and capacitor in series satisfy
$$ \frac qC + Ri = E(t), \qquad i = q', $$
so
$$ \boxed{R\frac{dq}{dt} + \frac qC = E(t).} $$
For positive constant $R, C$ and a constant source $E_0$,
$$ q(t) = CE_0 + [q(0) - CE_0]e^{-t/(RC)}. $$
The time constant is $\tau = RC$.

#### Simulation: the capacitor voltage cannot jump

Drive the same switched source through $R = 10\ \Omega$ into a capacitor,  
and integrate
$$ R\frac{dq}{dt} + \frac qC = E(t), \qquad q(0) = 0, $$
for two capacitances.  
The plot follows the capacitor voltage $q/C$ against the source.

```python
def rc_rhs(charge, t, C):
    return (lr_source(t) - float(charge[0])/C)/lr_R
rc_curves = [(C, color, odeint(rc_rhs, [0.0], lr_times, args=(float(C),), hmax=0.001)[:, 0]/float(C))
             for C, color in [(0.001, '#0072B2'), (0.005, '#D55E00')]]
```

```python
rc_plot = line([(t, lr_source(t)) for t in lr_times], color='silver',
               thickness=1.5, legend_label='Source $E(t)$')
for C, color, voltage in rc_curves:
    rc_plot += line(list(zip(lr_times, voltage)), color=color, thickness=2,
                    legend_label=rf'$C = {float(C)}$ F, $\tau = {float(C)*lr_R}$ s')
rc_plot.show(xmin=0, xmax=0.6, ymin=-0.6, ymax=15, figsize=[9, 4],
             axes_labels=['Time $t$ (s)', 'Capacitor voltage $q/C$ (V)'],
             frame=True, axes=False, legend_options={'loc': 'upper right'})
```

**Read the picture:** now the capacitor voltage is the continuous quantity,  
while the current $i = (E - q/C)/R$ jumps by $1.2$ A at every switching.  
The small capacitor, $\tau = 0.01$ s, charges fully inside each window;  
the large one, $\tau = 0.05$ s, reaches only $86\%$ of $12$ V.  
In the LR circuit the current was the smooth one and the inductor voltage jumped.

### LRC series circuit

![Series circuit (source PDF, p. 90)](0923/de2-circuit-56.png)

Kirchhoff's voltage law gives
$$ V_L + V_R + V_C = E(t). $$
The three component laws are
$$ V_L = L\frac{di}{dt}, \qquad V_R = Ri, \qquad V_C = \frac qC, $$
and current is the rate of change of charge: $i = dq/dt$.  
Therefore,
$$ \boxed{L\frac{d^2q}{dt^2} + R\frac{dq}{dt} + \frac qC = E(t).} $$

#### Why does the inductor create a second derivative?

An inductor responds to the **change of current**, not to charge itself:
$$ V_L = Li'. $$
But $i = q'$, so
$$ i' = q'' \qquad\Longrightarrow\qquad V_L = Lq''. $$
The second derivative appears only because the circuit equation is written in terms of charge.  
An LR circuit written in terms of current still has the first-order equation $Li' + Ri = E$.

The capacitor stores energy $q^2/(2C)$, while the inductor stores energy $Li^2/2$.  
Their natural state variables $q$ and $i$ can be prescribed independently,  
so an LRC circuit needs
$$ q(0) \qquad\text{and}\qquad i(0) = q'(0). $$
The resistor dissipates energy but stores no independent state.  
One storage element gives the first-order LR or RC model;  
two storage elements give the second-order LRC model.

#### Simulation: three damping regimes

Close the switch once at $t = 0$ with a constant $12$ V source,  
$L = 1$ H and $C = 0.01$ F, so that
$$ L\frac{d^2q}{dt^2} + R\frac{dq}{dt} + \frac qC = E_0, \qquad q(0) = q'(0) = 0. $$
Critical damping occurs at $R = 2\sqrt{L/C} = 20\ \Omega$;  
the simulation compares $R = 5$, $20$, and $60\ \Omega$.

```python
lrc_E0, lrc_L, lrc_C = float(12), float(1), float(0.01)
def lrc_rhs(state, t, R):
    charge, current = state
    return [current, (lrc_E0 - R*current - charge/lrc_C)/lrc_L]
lrc_times = np.linspace(0, 2, 2001)
lrc_curves = [(R, color, name, odeint(lrc_rhs, [0.0, 0.0], lrc_times, args=(float(R),))[:, 0])
              for R, color, name in [(5, '#0072B2', 'underdamped'), (20, '#009E73', 'critical'),
                                     (60, '#D55E00', 'overdamped')]]
```

```python
lrc_plot = line([(0, lrc_C*lrc_E0), (2, lrc_C*lrc_E0)], color='black',
                linestyle='--', alpha=0.6, legend_label='Steady charge $CE_0$')
for R, color, name, charge in lrc_curves:
    lrc_plot += line(list(zip(lrc_times, charge)), color=color, thickness=2,
                     legend_label=rf'$R = {R}\ \Omega$, {name}')
lrc_plot.show(xmin=0, xmax=2, ymin=0, ymax=0.2, figsize=[9, 4],
              axes_labels=['Time $t$ (s)', 'Charge $q$ (C)'],
              frame=True, axes=False, legend_options={'loc': 'lower right'})
```

**Read the picture:** all three settle at $CE_0 = 0.12$ C.  
Below critical resistance the charge overshoots to $0.173$ C and rings at about $10$ rad/s,  
which is $1/\sqrt{LC}$.  
At and above critical resistance the rise is monotone, and more resistance means slower.  
Oscillation needs the second derivative: the LR and RC circuits can only creep to their steady values.

### Example 7: switching on an LR circuit

Take $E = 12$ V, $L = 1/2$ H, $R = 10\ \Omega$, and $i(0) = 0$:
$$ \frac12 i' + 10i = 12 \quad\Longrightarrow\quad i' + 20i = 24. $$
Multiply by $e^{20t}$:
$$ (e^{20t}i)' = 24e^{20t}. $$
Integration gives $i = 6/5 + Ce^{-20t}$.

The initial current gives $C = -6/5$, hence
$$ \boxed{i(t) = \frac65(1 - e^{-20t})\ \mathrm A.} $$
The current approaches $1.2$ A and the time constant is $0.05$ s.

At switching, the inductor limits how fast current changes.  
At constant-source steady state, $i' = 0$ and the resistor sets $i = E/R$.

For an RC circuit under a constant source, charge approaches $CE_0$  
and current approaches zero.  
The resistor and capacitor together determine the charging timescale $RC$.

For either circuit, separate the decaying term from the long-term response.

### Reversible Reactions and Le Châtelier's Principle

Consider the reversible first-order reaction
$$ A \mathrel{\mathop{\rightleftharpoons}^{k_f}_{k_r}} B. $$
Let $x(t) = [B](t)$ and let the conserved total concentration be
$$ C = [A] + [B]. $$
Then $[A] = C - x$, so mass action gives
$$ \boxed{x' = k_f(C - x) - k_r x = k_fC - (k_f + k_r)x.} $$

The equilibrium concentration is
$$ x_* = \frac{k_f}{k_f + k_r}C, $$
and the deviation from equilibrium satisfies
$$ (x - x_*)' = -(k_f + k_r)(x - x_*). $$
Thus the reaction always moves exponentially toward the current equilibrium.

Suppose the system is at equilibrium and an amount $\Delta A$ is suddenly added.  
The concentration $x = [B]$ cannot jump, but the total $C$ and equilibrium $x_*$ do jump.  
The new equilibrium is higher by
$$ \Delta x_* = \frac{k_f}{k_f + k_r}\Delta A. $$
Immediately after the addition, $x' = k_f\Delta A > 0$:  
the reaction shifts toward $B$, consuming part of the added $A$.

```python
import numpy as np
kf, kr = 0.8, 0.4
t_add, C_before, added_A = 4.0, 1.0, 0.5
C_after = C_before + added_A
B_eq_before = kf*C_before/(kf + kr)
B_eq_after = kf*C_after/(kf + kr)
le_times = np.linspace(0, 10, 501)
le_B = np.where(le_times < t_add, B_eq_before,
                B_eq_after + (B_eq_before - B_eq_after)*np.exp(-(kf + kr)*(le_times - t_add)))
le_eq = np.where(le_times < t_add, B_eq_before, B_eq_after)
```

```python
le_plot = line(list(zip(le_times, le_B)), color='#0072B2', thickness=3, legend_label='$[B](t)$')
le_plot += line(list(zip(le_times, le_eq)), color='#D55E00', linestyle='--',
                thickness=2, legend_label='current equilibrium')
le_plot += line([(t_add, 0), (t_add, 1.1*B_eq_after)], color='gray', linestyle=':')
le_plot.show(xmin=0, xmax=10, ymin=0, ymax=1.1*B_eq_after, figsize=[9, 4],
             axes_labels=['Time $t$', 'Concentration'], frame=True, axes=False,
             legend_options={'loc': 'lower right'})
```

**Read the picture:** the orange equilibrium level jumps when $A$ is added.  
The blue concentration $[B]$ remains continuous, then relaxes toward the new equilibrium.  
Le Châtelier's principle is the sign of the restoring term made visible.

### From Scalar Linear Models to Higher-Dimensional States (hARD)

A scalar equation tracks one quantity.  
A population of interacting objects or a random system has a larger state.

Instead of solving for the entire state or probability distribution,  
we can track a few observables such as a mean, a second moment, or a variance.  
In favorable models, these observables form a closed linear system.

#### Random radioactive decay: mean, variance, and distribution

The exponential model describes an average amount.  
At the atomic level, the time at which an individual atom decays is random.

Start with exactly $N_0$ atoms.  
Let $N(t)$ be the random number still present at time $t$,  
and suppose each atom independently decays at rate $\lambda > 0$.

Given $N(t) = n$, during a short interval $\Delta t$,

- the probability of one decay is $\lambda n\,\Delta t + o(\Delta t)$;
- the probability of two or more decays is $o(\Delta t)$.

Thus
$$ \mathbb E[\Delta N \mid N = n] = -\lambda n\,\Delta t + o(\Delta t). $$

Let
$$ m(t) = \mathbb E[N(t)]. $$
Average the short-time change and divide by $\Delta t$:
$$ \boxed{m' = -\lambda m, \qquad m(0) = N_0.} $$
Therefore,
$$ \boxed{m(t) = N_0e^{-\lambda t}.} $$

To obtain the variance, first track the second moment
$$ q(t) = \mathbb E[N(t)^2]. $$
One decay changes $n^2$ by
$$ (n - 1)^2 - n^2 = -2n + 1. $$
Hence
$$ \boxed{q' = -2\lambda q + \lambda m.} $$

Let
$$ v(t) = \operatorname{Var}[N(t)] = q(t) - m(t)^2. $$
Differentiate and use the equations for $m$ and $q$:
$$ v' = q' - 2mm' = \lambda m - 2\lambda v. $$
Thus the mean and variance satisfy the triangular system
$$ \boxed{m' = -\lambda m, \qquad v' = \lambda m - 2\lambda v.} $$

Because the initial number is known exactly,
$$ m(0) = N_0, \qquad v(0) = 0. $$
Substitute $m(t) = N_0e^{-\lambda t}$ into the equation for $v$:
$$ v' + 2\lambda v = \lambda N_0e^{-\lambda t}. $$
The integrating factor $e^{2\lambda t}$ gives
$$ \boxed{v(t) = N_0\left(e^{-\lambda t} - e^{-2\lambda t}\right).} $$

For one atom, let $I_j(t) = 1$ if it survives until time $t$, and $0$ otherwise.  
Its survival probability solves
$$ p' = -\lambda p, \qquad p(0) = 1, $$
so $p(t) = e^{-\lambda t}$.  
Because the atoms decay independently, the variables $I_1, \ldots, I_{N_0}$  
are independent Bernoulli random variables with parameter $p$.

Since
$$ N(t) = I_1(t) + \cdots + I_{N_0}(t), $$
exactly $k$ survivors can be chosen in $\binom{N_0}{k}$ ways.  
Therefore,
$$ \mathbb P(N(t) = k) = \binom{N_0}{k}p^k(1 - p)^{N_0-k}, $$
and hence
$$ \boxed{N(t) \sim \operatorname{Binomial}(N_0, p).} $$

The probability generating function packages the entire distribution:
$$ G(z,t) = \mathbb E[z^{N(t)}] = (1 - p + pz)^{N_0}. $$
Its derivatives give every factorial moment:
$$ \boxed{\mathbb E[(N)_r] = \frac{\partial^r G}{\partial z^r}(1,t) = (N_0)_r p^r,} $$
where $(N)_r = N(N - 1)\cdots(N - r + 1)$.  
Thus the generating function, not merely the first two moments, identifies the distribution.

For $r = 1, 2$,
$$ \mathbb E[N] = N_0p, \qquad \mathbb E[N(N - 1)] = N_0(N_0 - 1)p^2. $$
Since $N^2 = N(N - 1) + N$, this recovers
$$ \operatorname{Var}(N) = N_0p(1 - p). $$
The variance is largest at $p = 1/2$, after exactly one half-life:
$$ \boxed{t = \frac{\ln2}{\lambda}, \qquad v_{\max} = \frac{N_0}{4}.} $$

#### Opinion consensus: a conserved mean and a decaying variance

Let $x_i(t)$ be the opinion of person $i$.  
Suppose everyone moves toward the current group average
$$ \bar{x} = \frac1n\sum_{i = 1}^n x_i $$
according to
$$ x_i' = k(\bar{x} - x_i), \qquad k > 0. $$
This is an $n$-dimensional linear system.

Average all $n$ equations:
$$ \bar{x}' = \frac{k}{n}\sum_{i = 1}^n(\bar{x} - x_i) = 0. $$
The group mean is conserved.  
No opinion is created or destroyed; it is only redistributed among the people.

Measure disagreement by
$$ V(t) = \frac1n\sum_{i = 1}^n(x_i - \bar{x})^2. $$
Because $\bar{x}' = 0$,
$$ V' = \frac2n\sum_{i = 1}^n(x_i - \bar{x})x_i' = -2kV. $$
Therefore,
$$ \boxed{\bar{x}(t) = \bar{x}(0), \qquad V(t) = V(0)e^{-2kt}.} $$

#### Random branching: a constant mean can hide growing uncertainty

Each individual gives birth at rate $b$ and dies at rate $d$.  
Given $N(t) = n$, the population changes to $n + 1$ at rate $bn$  
and to $n - 1$ at rate $dn$.

For $m(t) = \mathbb E[N(t)]$,
$$ \boxed{m' = (b - d)m.} $$

A birth changes $n^2$ by $2n + 1$, while a death changes it by $-2n + 1$.  
Thus, for $q(t) = \mathbb E[N(t)^2]$,
$$ q' = 2(b - d)q + (b + d)m. $$
Consequently, the variance satisfies
$$ \boxed{v' = 2(b - d)v + (b + d)m.} $$

In the critical case $b = d$ and with a known initial population $N_0$,
$$ m(t) = N_0, \qquad v' = 2bN_0. $$
Therefore,
$$ \boxed{v(t) = 2bN_0t.} $$
The expected population remains constant, even while the outcomes spread farther apart.  
Some populations become extinct while a few become very large.

#### Random reset: accumulation interrupted by sudden loss

Let $X(t)$ grow at constant rate $r$ between random reset events.  
During $\Delta t$, a reset sends $X$ to zero with probability
$$ \lambda\,\Delta t + o(\Delta t). $$
The state may represent the age of a work session,  
the length of a streak, or accumulated work since the last failure.

Let
$$ m(t) = \mathbb E[X(t)], \qquad q(t) = \mathbb E[X(t)^2]. $$
Continuous growth contributes $r$ to $m'$ and $2rm$ to $q'$.  
A reset removes the current value, giving
$$ \boxed{m' = r - \lambda m, \qquad q' = 2rm - \lambda q.} $$
Again, the two moments form a triangular linear system.

At equilibrium,
$$ m_\infty = \frac r\lambda, \qquad q_\infty = \frac{2r^2}{\lambda^2}. $$
Therefore,
$$ \boxed{\operatorname{Var}(X)_\infty = q_\infty - m_\infty^2 = \frac{r^2}{\lambda^2}.} $$
Faster accumulation increases both the mean and spread;  
more frequent resets decrease both.

#### Random queue: arrivals and independent completions

Jobs arrive at rate $a$.  
Each unfinished job is completed independently at rate $\mu$.  
If $N(t) = n$, the total completion rate is therefore $\mu n$.

This can model unfinished requests, unanswered messages,  
borrowed books, or molecules waiting to be removed.

For $m(t) = \mathbb E[N(t)]$, arrivals add $a$ and completions remove $\mu m$:
$$ \boxed{m' = a - \mu m.} $$
Using the changes $2n + 1$ and $-2n + 1$ in $n^2$ gives
$$ q' = a + (2a + \mu)m - 2\mu q. $$
Equivalently,
$$ \boxed{v' = a + \mu m - 2\mu v.} $$

At equilibrium,
$$ \boxed{m_\infty = \frac a\mu, \qquad v_\infty = \frac a\mu.} $$
The same scalar equation $m' = a - \mu m$ appeared in the mixing tank.  
Here it describes an expected count rather than a physical concentration.

#### Noisy mean reversion: decay balanced by new uncertainty

Suppose a random quantity $X(t)$ is pulled toward a baseline $\theta$.  
Over a short interval, assume
$$ \mathbb E[\Delta X \mid X = x] = -k(x - \theta)\,\Delta t + o(\Delta t), $$
$$ \operatorname{Var}(\Delta X \mid X = x) = \sigma^2\,\Delta t + o(\Delta t). $$
Equivalently, this is the Ornstein–Uhlenbeck SDE
$$ dX = -k(X - \theta)\,dt + \sigma\,dW. $$
Here $W$ is standard Brownian motion.  
The drift restores equilibrium, while the diffusion continually injects random variation.

For $m(t) = \mathbb E[X(t)]$ and $q(t) = \mathbb E[X(t)^2]$,
$$ m' = -k(m - \theta), $$
$$ q' = -2k(q - \theta m) + \sigma^2. $$
Subtracting the equation for $m^2$ gives the closed variance equation
$$ \boxed{v' = \sigma^2 - 2kv.} $$

The long-term mean and variance are
$$ \boxed{m_\infty = \theta, \qquad v_\infty = \frac{\sigma^2}{2k}.} $$
Without noise, every realization would converge to the same value.  
With noise, the mean still converges to $\theta$, but the distribution keeps a nonzero width.

For a known initial value $X(0) = x_0$,
$$ m(t) = \theta + (x_0 - \theta)e^{-kt}, $$
$$ v(t) = \frac{\sigma^2}{2k}(1 - e^{-2kt}), \qquad \operatorname{sd}[X(t)] = \sqrt{v(t)}. $$

##### Simulation: noisy paths around a restoring mean

```python
import numpy as np
rng = np.random.default_rng(int(3207))
theta, k, sigma, x0 = 2.0, 0.8, 1.2, 6.0
T_ou, dt_ou, number_of_ou_paths = 8.0, 0.01, 12
ou_times = np.linspace(0, T_ou, int(T_ou/dt_ou) + 1)
ou_paths = np.full((number_of_ou_paths, len(ou_times)), x0)
for j in range(len(ou_times) - 1):
    noise = rng.standard_normal(number_of_ou_paths)
    ou_paths[:, j + 1] = ou_paths[:, j] - k*(ou_paths[:, j] - theta)*dt_ou + sigma*np.sqrt(dt_ou)*noise
```

```python
ou_plot = Graphics()
for path in ou_paths:
    ou_plot += line(list(zip(ou_times, path)), color='#0072B2', alpha=0.45)
ou_mean = theta + (x0 - theta)*np.exp(-k*ou_times)
ou_variance = sigma**2/(2*k)*(1 - np.exp(-2*k*ou_times))
ou_one_sigma = np.sqrt(ou_variance)
```

```python
ou_plot += line(list(zip(ou_times, ou_mean)), color='black', thickness=3, legend_label='mean')
ou_plot += line(list(zip(ou_times, ou_mean + ou_one_sigma)), color='#009E73',
                linestyle='--', thickness=2, legend_label=r'mean $\pm$ one sigma')
ou_plot += line(list(zip(ou_times, ou_mean - ou_one_sigma)), color='#009E73',
                linestyle='--', thickness=2)
ou_plot += line([(0, theta), (T_ou, theta)], color='gray', linestyle=':')
ou_plot.show(xmin=0, xmax=T_ou, figsize=[9, 4],
             axes_labels=['Time $t$', '$X(t)$'], frame=True, axes=False,
             legend_options={'loc': 'upper right'})
```

**Read the picture:** the black mean relaxes from $x_0$ toward the gray baseline $\theta$.  
The green one-standard-deviation band starts with zero width and approaches $\sigma/\sqrt{2k}$.  
Individual paths fluctuate around the mean and can cross the band repeatedly.

#### Random switching: a two-dimensional system reduced by conservation

A system switches from off to on at rate $\alpha$,  
and from on to off at rate $\beta$.  
Let $p_0(t)$ and $p_1(t)$ be the two state probabilities.

Probability flows between the states according to
$$ p_0' = -\alpha p_0 + \beta p_1, \qquad p_1' = \alpha p_0 - \beta p_1. $$

Because $p_0 + p_1 = 1$, set $p = p_1$ to obtain
$$ \boxed{p' = \alpha(1 - p) - \beta p.} $$
Thus
$$ p_\infty = \frac{\alpha}{\alpha + \beta}. $$
For the binary state $X \in \{0, 1\}$,
$$ \mathbb E[X] = p, \qquad \operatorname{Var}(X) = p(1 - p). $$

#### Insurance surplus: what moments can and cannot answer

An insurer starts with surplus $u$ and receives premiums at rate $c$.  
Claims arrive at rate $\lambda$, and the independent claim sizes are $Y_1, Y_2, \ldots$.  
The classical surplus process is
$$ U(t) = u + ct - \sum_{j = 1}^{N(t)}Y_j. $$
Write
$$ \mu_1 = \mathbb E[Y], \qquad \mu_2 = \mathbb E[Y^2]. $$

During $\Delta t$, premium income contributes $c\,\Delta t$,  
while one claim occurs with probability $\lambda\,\Delta t + o(\Delta t)$.  
Therefore,
$$ \boxed{m' = c - \lambda\mu_1, \qquad v' = \lambda\mu_2,} $$
where $m(t) = \mathbb E[U(t)]$ and $v(t) = \operatorname{Var}[U(t)]$.

For a known initial surplus,
$$ m(t) = u + (c - \lambda\mu_1)t, \qquad v(t) = \lambda\mu_2t. $$
The condition $c > \lambda\mu_1$ gives positive expected drift.  
It does not guarantee survival: a sufficiently large early claim can still cause ruin.  
Mean and variance alone do not determine the ruin probability.

A diffusion approximation keeps the same local mean and variance rates:
$$ dU = a\,dt + \sigma\,dW, \qquad a = c - \lambda\mu_1, \qquad \sigma^2 = \lambda\mu_2. $$
For $a > 0$, its eventual ruin probability $\psi(u)$ solves
$$ \frac{\sigma^2}{2}\psi'' + a\psi' = 0, \qquad \psi(0) = 1, \qquad \psi(\infty) = 0, $$
and therefore
$$ \boxed{\psi(u) = e^{-2au/\sigma^2}.} $$

For the diffusion model with known initial surplus $u_0$,  
$$ m(t) = \mathbb E[U(t)] = u_0 + at, $$
$$ \operatorname{Var}[U(t)] = \sigma^2t, \qquad \operatorname{sd}[U(t)] = \sigma\sqrt t. $$
The curves $m(t) - \sigma\sqrt t$ and $m(t) + \sigma\sqrt t$ form the pointwise one-standard-deviation band.  
At each fixed time, the band contains about $68\%$ of the Gaussian distribution.  
It is not a pair of barriers that traps an entire sample path.

##### Simulation: positive drift does not prevent ruin

```python
import numpy as np
rng = np.random.default_rng(int(1007))
u0, drift, volatility = 4.0, 0.5, 2.0
T, dt, number_of_paths = 12.0, 0.01, 16
risk_times = np.linspace(0, T, int(T/dt) + 1)
increments = drift*dt + volatility*np.sqrt(dt)*rng.standard_normal((number_of_paths, len(risk_times) - 1))
risk_paths = u0 + np.cumsum(increments, axis=1)
risk_paths = np.column_stack([np.full(number_of_paths, u0), risk_paths])
was_ruined = np.min(risk_paths, axis=1) <= 0
```

```python
risk_plot = Graphics()
seen = {False: False, True: False}
for path, ruined in zip(risk_paths, was_ruined):
    end = np.argmax(path <= 0) + 1 if ruined else len(path)
    label = ('ruin' if ruined else 'survives') if not seen[bool(ruined)] else None
    options = {'legend_label': label} if label else {}
    risk_plot += line(list(zip(risk_times[:end], path[:end])),
                      color='#D55E00' if ruined else '#0072B2', alpha=0.55, **options)
    seen[bool(ruined)] = True
```

```python
risk_mean = u0 + drift*risk_times
risk_one_sigma = volatility*np.sqrt(risk_times)
risk_plot += line([(0, 0), (T, 0)], color='gray', thickness=2)
risk_plot += line(list(zip(risk_times, risk_mean)), color='black',
                  thickness=3, legend_label='mean')
risk_plot += line(list(zip(risk_times, risk_mean + risk_one_sigma)), color='#009E73',
                  linestyle='--', thickness=2, legend_label=r'mean $\pm$ one sigma')
risk_plot += line(list(zip(risk_times, risk_mean - risk_one_sigma)), color='#009E73',
                  linestyle='--', thickness=2)
risk_plot.show(xmin=0, xmax=T, ymin=-0.5, figsize=[9, 4],
               axes_labels=['Time $t$', 'Surplus $U(t)$'], frame=True, axes=False,
               legend_options={'loc': 'upper left'})
```

**Read the picture:** the solid black curve is the mean, and the green dashed curves form its pointwise one-standard-deviation band.  
The gray horizontal line is the ruin boundary.  
The band does not trap individual paths; a path can cross it repeatedly.  
Although the mean rises because the drift is positive, the orange paths still hit zero.  
Ruin is a first-passage event, not a question about the expected value at a fixed time.

#### One linear structure, several meanings

The equation $y' = a - by$ can describe very different observables:

| Observable $y$ | Input $a$ | Removal or restoration $b$ |
|---|---:|---:|
| salt amount | salt inflow | fractional outflow |
| expected unfinished jobs | arrival rate | completion rate |
| probability of being on | off-to-on rate | sum of switching rates |
| mean of a noisy variable | $k\theta$ | restoring rate $k$ |

The main modeling decision is not how to solve the ODE.  
It is what the state means and which transitions change it.

Once those choices are made, conservation, expectation, and moment identities  
often reduce a high-dimensional system to a small collection of linear equations.

## 3.2 Nonlinear Models

### The logistic equation

Exponential growth assumes a constant per-capita growth rate.  
To account for limited resources, let that rate decrease with population:
$$ \boxed{P' = P(a - bP) = aP\left(1 - \frac PK\right), \qquad K = \frac ab.} $$
Assume $a > 0$, $b > 0$.  
The number $K$ is the **carrying capacity**.

The equilibria are $P = 0$ and $P = K$.

- If $0 < P < K$, then $P' > 0$.
- If $P > K$, then $P' < 0$.
- Near $P = 0$, a small positive population grows away from zero.

Thus $K$ is stable and $0$ is unstable.  
For $P_0 > 0$, the solution approaches $K$ as $t \to \infty$.

![Logistic solution curves (source PDF, p. 94)](0923/de2-logistic-60.png)

### Solving the logistic equation

For a non-equilibrium solution, separate:
$$ \frac{dP}{P(a - bP)} = dt. $$
Use
$$ \frac{1}{P(a - bP)} = \frac{1/a}{P} + \frac{b/a}{a - bP}. $$
After integration,
$$ \frac1a\ln|P| - \frac1a\ln|a - bP| = t + C. $$

Exponentiating gives
$$ \frac{P}{a - bP} = C_1e^{at}. $$
With $P(0) = P_0$, the result can be written as
$$ \boxed{P(t) = \frac{aP_0}{bP_0 + (a - bP_0)e^{-at}}.} $$
For $P_0 > 0$, this is equivalently
$$ \boxed{P(t) = \frac{K}{1 + (K/P_0 - 1)e^{-at}}.} $$
The first form also includes $P_0 = 0$; both include $P_0 = K$ where defined.

Differentiate the rate to locate an inflection point:
$$ P'' = (a - 2bP)P'. $$
A growing solution with $0 < P < K$ changes concavity at $P = K/2$.  
If $P_0 > K/2$, that point lies before $t = 0$.

A solution beginning above $K$ decreases toward $K$.

### Example 1: spread of a virus on a campus

A closed campus contains 1000 students. Initially one is infected.  
Assume the spread rate is proportional to both the infected population  
and the population not yet infected, with no recovery during the modeled period.

Let $x(t)$ be the infected count, with $t$ in days:
$$ x' = kx(1000 - x), \qquad x(0) = 1, \qquad x(4) = 50. $$
Find $x(6)$.

Use the logistic formula with $K = 1000$:
$$ x(t) = \frac{1000}{1 + 999e^{-1000kt}}. $$
The observation after four days gives
$$ 50 = \frac{1000}{1 + 999e^{-4000k}}, $$
$$ e^{-4000k} = \frac{19}{999}, \qquad 1000k = \frac14\ln\frac{999}{19}. $$

Therefore,
$$ \boxed{x(t) = \frac{1000}{1 + 999e^{-rt}}, \qquad r = \frac14\ln\frac{999}{19} \approx 0.9906\ \mathrm{day}^{-1}.} $$
At six days,
$$ \boxed{x(6) \approx 276\text{ students}.} $$
The model treats population as continuous; round only when reporting a count.

```python
import math
r = math.log(999 / 19) / 4
infected = lambda days: 1000 / (1 + 999 * math.exp(-r * days))
print(f'r = {r:.6f} per day')
print(f'x(0) = {infected(0):.0f}, x(4) = {infected(4):.0f}')
print(f'x(6) = {infected(6):.3f} students')
```

### Variants of the logistic model

| Model | Interpretation |
|---|---|
| $P' = P(a - bP) + h$ | Constant immigration |
| $P' = P(a - bP) - h$ | Constant emigration or harvesting |
| $P' = P(a - bP) - cP$ | Emigration proportional to population |
| $P' = P(a - bP) + ce^{-kP}$ | Immigration decreases as population rises |

For each model, find equilibria from $P' = 0$ and inspect the sign of $P'$.

### The Gompertz equation

Another limited-growth model is
$$ P' = P(a - b\ln P) = bP\ln\frac KP, \qquad K = e^{a/b}, \qquad P > 0. $$
The per-capita growth rate is proportional to $\ln(K/P)$.

Set $u = \ln P$. Then
$$ u' = a - bu, $$
which is linear even though the original equation is nonlinear.

For $b > 0$ and $P(0) = P_0 > 0$,
$$ u(t) = \frac ab + \left(\ln P_0 - \frac ab\right)e^{-bt}, $$
so
$$ \boxed{P(t) = K\exp\left[\ln\left(\frac{P_0}{K}\right)e^{-bt}\right].} $$
The population approaches $K$.

### Chemical reaction rates

Consider a reaction $A + B \to C$.  
Let $x(t)$ be the amount of product formed, with $x(0) = 0$.

- Initially there are amounts $a$ of $A$ and $b$ of $B$.
- Making one unit of product consumes $s_1$ units of $A$ and $s_2$ units of $B$.
- The remaining reactant amounts are $a - s_1x$ and $b - s_2x$.
- Assume constant volume and a rate proportional to their product.

The model is
$$ \boxed{\frac{dx}{dt} = k(a - s_1x)(b - s_2x), \qquad k > 0.} $$
It is separable:
$$ \int \frac{dx}{(a - s_1x)(b - s_2x)} = kt + C. $$
The physically meaningful range is
$$ 0 \le x \le \min\left(\frac a{s_1}, \frac b{s_2}\right), $$
assuming positive initial amounts and positive consumption coefficients.

If $D = as_2 - bs_1 \ne 0$, partial fractions give
$$ \frac1D\ln\left|\frac{a - s_1x}{b - s_2x}\right| = kt + C. $$
If $D = 0$, write $\alpha = a/s_1 = b/s_2$ instead:
$$ x' = ks_1s_2(\alpha - x)^2. $$
Each case can be integrated, and the limiting reactant determines  
the maximum amount of product.

### Example 2: second-order chemical reaction

A compound $C$ forms when chemicals $A$ and $B$ react.  
For every 1 g of $A$ consumed, 4 g of $B$ is consumed.  
Initially there are 50 g of $A$ and 32 g of $B$.

After 10 minutes, 30 g of $C$ has formed.  
Find the amount $x(t)$ of $C$, compute $x(15)$, and describe the long-term result.

Because the consumed masses occur in the ratio $1:4$,  
$x$ grams of $C$ uses $x/5$ grams of $A$ and $4x/5$ grams of $B$.

The remaining amounts are therefore
$$ 50 - \frac x5 \qquad\text{and}\qquad 32 - \frac{4x}{5}. $$

The law of mass action gives
$$ x' = k\left(50 - \frac x5\right)\left(32 - \frac{4x}{5}\right). $$
Absorb the constant factor into $k$ and write
$$ \boxed{x' = k(250 - x)(40 - x), \qquad x(0) = 0, \qquad x(10) = 30.} $$
The physical range is $0 \le x \le 40$.

Use the partial-fraction decomposition
$$ \frac{1}{(250 - x)(40 - x)}
= -\frac{1}{210(250 - x)} + \frac{1}{210(40 - x)}. $$
After integration,
$$ \ln\left(\frac{250 - x}{40 - x}\right) = rt + C, \qquad r = 210k. $$

The condition $x(0) = 0$ gives
$$ \frac{250 - x}{40 - x} = \frac{25}{4}e^{rt}. $$
Using $x(10) = 30$,
$$ 22 = \frac{25}{4}e^{10r}, \qquad
r = \frac1{10}\ln\frac{88}{25} \approx 0.1258\ \mathrm{min}^{-1}. $$

Solving for $x$ gives
$$ \boxed{x(t) = 1000\frac{1 - e^{-rt}}{25 - 4e^{-rt}}, \qquad r \approx 0.1258.} $$
Therefore,
$$ \boxed{x(15) \approx 34.78\ \mathrm g.} $$

As $t \to \infty$,
$$ \boxed{x(t) \to 40\ \mathrm g.} $$
Chemical $B$ is the limiting reactant: none remains,  
while $50 - 40/5 = 42$ g of $A$ remains.

### Reaction Order Is Not a Head Count

For an elementary reaction
$$ 2A + B \longrightarrow C, $$
mass action gives the third-order rate law
$$ r = k[A]^2[B]. $$
If $x$ is the reaction extent, then $[A] = a - 2x$ and $[B] = b - x$, so
$$ \boxed{x' = k(a - 2x)^2(b - x).} $$

Stoichiometric coefficients determine how concentrations change with $x$.  
Reaction order is determined by the exponents in the rate law.  
The two agree automatically only for an elementary reaction;  
an experimentally observed overall reaction may have a different rate law.

For the simpler model $A' = -kA^n$ with $A(0) = A_0$,
$$ A(t) = A_0e^{-kt} \qquad (n = 1), $$
$$ A(t) = \frac{A_0}{1 + kA_0t} \qquad (n = 2), $$
$$ A(t) = \frac{A_0}{\sqrt{1 + 2kA_0^2t}} \qquad (n = 3). $$
Higher-order decay becomes much slower after the concentration becomes small.

#### Simulation: first-, second-, and third-order decay

```python
import numpy as np
order_times = np.linspace(0, 6, 301)
A0, reaction_k = 1.0, 1.0
order_curves = {
    1: A0*np.exp(-reaction_k*order_times),
    2: A0/(1 + reaction_k*A0*order_times),
    3: A0/np.sqrt(1 + 2*reaction_k*A0**2*order_times),
}
```

```python
order_plot = Graphics()
order_colors = {1: '#0072B2', 2: '#009E73', 3: '#D55E00'}
for reaction_order, values in order_curves.items():
    order_plot += line(list(zip(order_times, values)), color=order_colors[reaction_order],
                       thickness=3, legend_label=f'order {reaction_order}')
order_plot.show(xmin=0, xmax=6, ymin=0, ymax=1, figsize=[9, 4],
                axes_labels=['Time $t$', 'Concentration $A(t)$'],
                frame=True, axes=False, legend_options={'loc': 'upper right'})
```

**Read the picture:** with $A_0 = k = 1$, all three curves begin with the same slope.  
Once $A < 1$, the factor $A^n$ is smaller for larger $n$.  
The third-order reaction therefore develops the longest tail.

### Two Protons Do Not Determine Reaction Order

Sulfuric acid is diprotic, so its two protons dissociate in two stages:
$$ \mathrm{H_2SO_4} \rightleftharpoons \mathrm{H^+} + \mathrm{HSO_4^-}, $$
$$ \mathrm{HSO_4^-} \rightleftharpoons \mathrm{H^+} + \mathrm{SO_4^{2-}}. $$
The first dissociation is much stronger than the second.  
Two ionizable protons mean two coupled equilibria, not a second- or third-order rate law.

For the second stage, an elementary kinetic description would use
$$ r_f = k_f[\mathrm{HSO_4^-}], \qquad
r_r = k_r[\mathrm{H^+}][\mathrm{SO_4^{2-}}]. $$
The forward term is first-order and the reverse term is second-order.  
The number of acidic hydrogens determines how many species must be tracked;  
the rate law determines reaction order.
