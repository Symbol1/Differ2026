In 4.1 we learned what to look for:  
$n$ independent solutions of $L y = 0$, plus one particular solution of $L y = g$.  
When the coefficients are constant, both can be found by algebra.

- 4.3: the homogeneous equation, through the roots of a polynomial.
- 4.4: a particular solution, by guessing its form from $g$.
- 4.5: the same guess, explained by operators that annihilate $g$.

## 4.3 Homogeneous Linear Equations with Constant Coefficients

### Exponentials are eigenfunctions of $D$

For a first-order equation $ay' + by = 0$, try $y = e^{mx}$:
$$ ame^{mx} + be^{mx} = (am + b)e^{mx} = 0. $$
Since $e^{mx} \ne 0$, the exponent must satisfy $am + b = 0$.  
The same idea works for every order, because
$$ D e^{mx} = me^{mx}, \qquad D^k e^{mx} = m^k e^{mx}. $$

Let $P(m) = a_n m^n + \cdots + a_1 m + a_0$ have real constant coefficients,  
and write the equation as $P(D) y = 0$. Then
$$ \boxed{P(D) e^{mx} = P(m) e^{mx}.} $$
So $e^{mx}$ solves $P(D) y = 0$ exactly when $P(m) = 0$.  
The equation $P(m) = 0$ is the **auxiliary equation**, also called the characteristic equation.  
In the language of linear algebra, $e^{mx}$ is an eigenfunction of $D$ with eigenvalue $m$.

### Second-order equations

For $ay'' + by' + cy = 0$ the auxiliary equation is
$$ am^2 + bm + c = 0, \qquad m = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a}. $$
The sign of the discriminant $b^2 - 4ac$ separates three cases.

#### Case I: distinct real roots

If $b^2 - 4ac > 0$, the roots $m_1 \ne m_2$ are real, and
$$ \boxed{y = c_1 e^{m_1 x} + c_2 e^{m_2 x}.} $$
The two exponentials are independent, since
$$ W(e^{m_1 x}, e^{m_2 x}) = (m_2 - m_1) e^{(m_1 + m_2)x} \ne 0. $$

#### Case II: a repeated real root

If $b^2 - 4ac = 0$, there is only one root $m_1 = -b/(2a)$ and only one exponential.  
Reduction of order produced a second solution in Example 3 of 4.2:
$$ \boxed{y = c_1 e^{m_1 x} + c_2 xe^{m_1 x}.} $$
Below we will see a shorter reason why the factor $x$ appears.

#### Case III: complex conjugate roots

If $b^2 - 4ac < 0$, the roots are $m = \alpha \pm i\beta$ with $\beta > 0$.  
Euler's formula $e^{i\theta} = \cos\theta + i\sin\theta$ gives the complex solution
$$ e^{(\alpha + i\beta)x} = e^{\alpha x} \cos(\beta x) + ie^{\alpha x} \sin(\beta x). $$
Because $P$ has real coefficients, $P(D)(u + iv) = P(D) u + iP(D) v$,  
so the real and imaginary parts are real solutions on their own:
$$ \boxed{y = e^{\alpha x}\bigl(c_1 \cos(\beta x) + c_2 \sin(\beta x)\bigr).} $$

The real part $\alpha$ is a growth or decay rate, and the imaginary part $\beta$ is an angular frequency.  
The conjugate root $\alpha - i\beta$ adds nothing new:  
its real and imaginary parts are the same two functions, up to sign.

### Example 1: the three cases

| Equation | Auxiliary equation | Roots | General solution |
|---|---|---|---|
| $2y'' - 5y' - 3y = 0$ | $(2m + 1)(m - 3) = 0$ | $-\frac12$, $3$ | $c_1 e^{-x/2} + c_2 e^{3x}$ |
| $y'' - 10y' + 25y = 0$ | $(m - 5)^2 = 0$ | $5$, $5$ | $c_1 e^{5x} + c_2 xe^{5x}$ |
| $y'' + 4y' + 7y = 0$ | $m^2 + 4m + 7 = 0$ | $-2 \pm \sqrt3 i$ | $e^{-2x}\bigl(c_1 \cos(\sqrt3 x) + c_2 \sin(\sqrt3 x)\bigr)$ |

### Example 2: an initial value problem

Solve
$$ 4y'' + 4y' + 17y = 0, \qquad y(0) = -1, \qquad y'(0) = 2. $$
The roots of $4m^2 + 4m + 17 = 0$ are $m = -\frac12 \pm 2i$, so
$$ y = e^{-x/2}\bigl(c_1 \cos(2x) + c_2 \sin(2x)\bigr). $$
Then $y(0) = c_1 = -1$ and $y'(0) = -\frac12 c_1 + 2c_2 = 2$, so $c_2 = \frac34$:
$$ \boxed{y = e^{-x/2}\left(-\cos(2x) + \frac34 \sin(2x)\right).} $$
The solution oscillates with angular frequency $2$ while its amplitude decays like $e^{-x/2}$.

### Two equations worth knowing

For $k > 0$,
$$ y'' + k^2 y = 0 \quad\Longrightarrow\quad y = c_1 \cos(kx) + c_2 \sin(kx), $$
$$ y'' - k^2 y = 0 \quad\Longrightarrow\quad y = c_1 e^{kx} + c_2 e^{-kx} = C_1 \cosh(kx) + C_2 \sinh(kx). $$
In initial-data form, as in the 0923 homework,
$$ y = y(0) \cos(kx) + \frac{y'(0)}{k}\sin(kx), \qquad y = y(0) \cosh(kx) + \frac{y'(0)}{k}\sinh(kx). $$
On a finite interval $\cosh$ and $\sinh$ are often more convenient; as $x \to \infty$ the exponentials are.

#### Picture: roots in the plane, shapes on the line

Follow
$$ y'' + by' + y = 0, \qquad y(0) = 1, \qquad y'(0) = 0 $$
for several damping coefficients $b \ge 0$.  
The roots satisfy $m_1 + m_2 = -b$ and $m_1 m_2 = 1$.  
For distinct roots, real or complex, the solution is
$$ y = \frac{m_2 e^{m_1 t} - m_1 e^{m_2 t}}{m_2 - m_1}, $$
and for the double root $m = -1$ at $b = 2$ it is $y = (1 + t)e^{-t}$.

```python
import numpy as np
locus_cases = [(0.4, '#0072B2', 'underdamped'), (2, '#009E73', 'critical'), (5, '#D55E00', 'overdamped')]
locus_times = np.linspace(0, 12, 601)
def damping_roots(b):
    return np.roots([1.0, float(b), 1.0]).astype(complex)
def damped_solution(b, t):
    if b == 2:
        return (1 + t)*np.exp(-t)
    m1, m2 = damping_roots(b)
    return ((m2*np.exp(m1*t) - m1*np.exp(m2*t))/(m2 - m1)).real
```

```python
roots_plot = circle((0, 0), 1, color='gray', linestyle=':')
for b in srange(0, 5.01, 0.1):                  # the path of the roots as b grows
    roots_plot += points([(float(r.real), float(r.imag)) for r in damping_roots(b)], color='silver', size=8)
for b, color, name in locus_cases:
    roots_plot += points([(float(r.real), float(r.imag)) for r in damping_roots(b)], color=color, size=60,
                         zorder=5, legend_label=f'$b = {float(b):g}$')
roots_plot.show(aspect_ratio=1, xmin=-5, xmax=0.5, ymin=-1.3, ymax=1.3, figsize=[9, 3.5], frame=True,
                axes=True, axes_labels=[r'$\mathrm{Re}\,m$', r'$\mathrm{Im}\,m$'],
                legend_options={'loc': 'upper left'})
```

```python
shape_plot = line([(0, 0), (12, 0)], color='gray', linestyle=':')
for b, color, name in locus_cases:
    shape_plot += line(list(zip(locus_times, damped_solution(b, locus_times))), color=color, thickness=2,
                       legend_label=f'$b = {float(b):g}$, {name}')
shape_plot.show(xmin=0, xmax=12, figsize=[9, 4], frame=True, axes=False,
                axes_labels=['$t$', '$y$'], legend_options={'loc': 'upper right'})
```

**Read the picture:** for $0 < b < 2$ the roots are a conjugate pair on the unit circle, since $m_1 m_2 = 1$.  
As $b$ grows they slide to $-1$, meet at $b = 2$, and then split along the real axis.  
The real part is the decay rate: the blue solution oscillates at $\beta \approx 0.98$ inside the envelope $e^{-0.2t}$.  
For $b = 5$ one root is near $-0.21$, so the orange solution creeps back slowly.  
More damping is not always faster: the critical case $b = 2$ settles first, as in the LRC simulation of 1007.

### Higher-order equations

For $P(D) y = 0$ of order $n$, factor the auxiliary polynomial $P(m)$ completely.  
Each root contributes solutions:

| Root of $P(m) = 0$ | Solutions |
|---|---|
| real $m$ of multiplicity $k$ | $e^{mx}, xe^{mx}, \ldots, x^{k - 1} e^{mx}$ |
| complex pair $\alpha \pm i\beta$ of multiplicity $k$ | $x^j e^{\alpha x} \cos(\beta x)$ and $x^j e^{\alpha x} \sin(\beta x)$, $j = 0, \ldots, k - 1$ |

Together these are $n$ independent solutions: a fundamental set.

### Why repeated roots give powers of $x$

The product rule gives the **exponential shift**
$$ (D - m)\bigl[e^{mx} f(x)\bigr] = me^{mx} f + e^{mx} f' - me^{mx} f = e^{mx} f'(x). $$
Applying it $k$ times,
$$ \boxed{(D - m)^k \bigl[e^{mx} f(x)\bigr] = e^{mx} f^{(k)}(x).} $$
If $(m - m_1)^k$ divides $P(m)$, then $P(D) = Q(D)(D - m_1)^k$,  
and every $x^j e^{m_1 x}$ with $j < k$ is annihilated, because $D^k x^j = 0$.

More generally, $D[e^{mx} f] = e^{mx}(D + m) f$, and applying this repeatedly gives
$$ \boxed{P(D)\bigl[e^{mx} f\bigr] = e^{mx} P(D + m) f} $$
for every polynomial $P$.  
This **shift rule** turns a question about $e^{mx} f$ into a question about $f$ alone.  
It will explain the multiplication rule of 4.4 and the annihilators of 4.5.

### Example 3: a cubic with an obvious root

For $y''' + 3y'' - 4y = 0$, the auxiliary polynomial $m^3 + 3m^2 - 4$ vanishes at $m = 1$.  
Dividing by $m - 1$,
$$ m^3 + 3m^2 - 4 = (m - 1)(m^2 + 4m + 4) = (m - 1)(m + 2)^2. $$
Hence
$$ \boxed{y = c_1 e^x + c_2 e^{-2x} + c_3 xe^{-2x}.} $$

### Example 4: a repeated complex pair

For $y^{(4)} + 2y'' + y = 0$, the auxiliary equation is $(m^2 + 1)^2 = 0$, so $\pm i$ are double roots:
$$ \boxed{y = c_1 \cos x + c_2 \sin x + c_3 x\cos x + c_4 x\sin x.} $$

### Finding roots by hand

Without a calculator, try these in order.

1. Factor out a power of $m$ when the lowest derivatives are missing, e.g. $m^5 + m^3 = m^3(m^2 + 1)$.
2. Look for structure: a perfect square, a quadratic in $m^2$, or $m^n = c$.
3. Test rational roots $p/q$, where $p$ divides $a_0$ and $q$ divides $a_n$; divide out each root found.
4. Finish the remaining quadratic with the formula.

### Example 5: a rational root

For $3y''' + 5y'' + 10y' - 4y = 0$, the candidates are $\pm1, \pm2, \pm4, \pm\frac13, \pm\frac23, \pm\frac43$.  
Testing them, $m = \frac13$ works, and division gives
$$ 3m^3 + 5m^2 + 10m - 4 = (3m - 1)(m^2 + 2m + 4). $$
The quadratic has roots $m = -1 \pm \sqrt3 i$, so
$$ \boxed{y = c_1 e^{x/3} + e^{-x}\bigl(c_2 \cos(\sqrt3 x) + c_3 \sin(\sqrt3 x)\bigr).} $$

### Example 6: $y^{(4)} + y = 0$

The roots of $m^4 = -1$ are the fourth roots of $-1 = e^{i\pi}$:
$$ m = e^{i(\pi + 2k\pi)/4} = \frac{\pm1 \pm i}{\sqrt2}. $$
Without polar form, complete the square instead:
$$ m^4 + 1 = (m^2 + 1)^2 - 2m^2 = (m^2 - \sqrt2 m + 1)(m^2 + \sqrt2 m + 1). $$
Therefore
$$ y = e^{x/\sqrt2}\left(c_1 \cos\frac{x}{\sqrt2} + c_2 \sin\frac{x}{\sqrt2}\right)
+ e^{-x/\sqrt2}\left(c_3 \cos\frac{x}{\sqrt2} + c_4 \sin\frac{x}{\sqrt2}\right). $$

### Initial value problems of higher order

For an $n$th-order problem, the $n$ initial conditions give $n$ linear equations for $c_1, \ldots, c_n$.  
Their coefficient matrix is the Wronskian matrix at $x_0$, which is invertible.  
For example, $y''' + 3y'' - 4y = 0$ with $y(0) = 1$, $y'(0) = y''(0) = 0$ requires
$$ c_1 + c_2 = 1, \qquad c_1 - 2c_2 + c_3 = 0, \qquad c_1 + 4c_2 - 4c_3 = 0, $$
so $c_1 = \frac49$, $c_2 = \frac59$, $c_3 = \frac23$, and
$$ y = \frac49 e^x + \frac59 e^{-2x} + \frac23 xe^{-2x}. $$

## 4.4 Undetermined Coefficients: Superposition Approach

### Guess the form, then match coefficients

To solve $P(D) y = g(x)$ we need $y_c$, from 4.3, and one particular solution $y_p$.  
When $g$ is built from polynomials, exponentials, sines, and cosines,  
its derivatives stay inside a finite family of functions.  
So we try $y_p$ as a combination of that family with **undetermined coefficients**,  
substitute, and match coefficients.  
The method fails for inputs such as $\ln x$, $1/x$, or $\tan x$; those need variation of parameters in 4.6.

### Example 1: a polynomial input

Solve $y'' + 4y' - 2y = 2x^2 - 3x + 6$.  
The roots of $m^2 + 4m - 2 = 0$ are $-2 \pm \sqrt6$, so $y_c = c_1 e^{(-2 + \sqrt6)x} + c_2 e^{(-2 - \sqrt6)x}$.  
Try $y_p = Ax^2 + Bx + C$:
$$ 2A + 4(2Ax + B) - 2(Ax^2 + Bx + C) = -2Ax^2 + (8A - 2B)x + (2A + 4B - 2C). $$

Matching coefficients with $2x^2 - 3x + 6$,
$$ -2A = 2, \qquad 8A - 2B = -3, \qquad 2A + 4B - 2C = 6, $$
so $A = -1$, $B = -\frac52$, $C = -9$, and
$$ \boxed{y = c_1 e^{(-2 + \sqrt6)x} + c_2 e^{(-2 - \sqrt6)x} - x^2 - \frac52 x - 9.} $$
The trial contains all lower powers, because differentiating $x^2$ produces $x$ and $1$.

### Example 2: a sine input needs a cosine too

Find a particular solution of $y'' - y' + y = 2\sin(3x)$.  
Differentiating $\sin(3x)$ produces $\cos(3x)$, so try $y_p = A\cos(3x) + B\sin(3x)$:
$$ (-8A - 3B)\cos(3x) + (3A - 8B)\sin(3x) = 2\sin(3x). $$
Then $-8A - 3B = 0$ and $3A - 8B = 2$, so
$$ \boxed{y_p = \frac{6}{73}\cos(3x) - \frac{16}{73}\sin(3x).} $$

### Example 3: superposition of inputs

Solve $y'' - 2y' - 3y = 4x - 5 + 6xe^{2x}$, where $y_c = c_1 e^{-x} + c_2 e^{3x}$.  
Split the input and find a particular solution for each piece, as in 4.1.

- For $4x - 5$, try $Ax + B$: $-2A - 3(Ax + B) = 4x - 5$ gives $A = -\frac43$, $B = \frac{23}{9}$.
- For $6xe^{2x}$, try $(Cx + E)e^{2x}$: matching gives $C = -2$, $E = -\frac43$.

Therefore
$$ \boxed{y = c_1 e^{-x} + c_2 e^{3x} - \frac43 x + \frac{23}{9} - \left(2x + \frac43\right)e^{2x}.} $$
The second computation is shorter with the shift rule:  
$P(D)[e^{2x} f] = e^{2x} P(D + 2) f$ with $P(D + 2) = D^2 + 2D - 3$,  
so $(D^2 + 2D - 3)(Cx + E) = 2C - 3Cx - 3E = 6x$.

### Example 4: a glitch

Find a particular solution of $y'' - 5y' + 4y = 8e^x$.  
The natural trial $Ae^x$ fails: $P(1) = 1 - 5 + 4 = 0$, so $P(D)(Ae^x) = 0$ for every $A$.  
The input already solves the homogeneous equation, since $y_c = c_1 e^x + c_2 e^{4x}$.  
Try $Axe^x$ instead. The shift rule gives
$$ P(D)(Axe^x) = e^x P(D + 1)(Ax) = e^x(D^2 - 3D)(Ax) = -3Ae^x, $$
so $A = -\frac83$ and $y_p = -\frac83 xe^x$.

### Trial forms

| Input $g(x)$ | Trial $y_p$ |
|---|---|
| $5$ | $A$ |
| $3x^2 - 2$ | $Ax^2 + Bx + C$ |
| $e^{5x}$ | $Ae^{5x}$ |
| $(9x - 2)e^{5x}$ | $(Ax + B)e^{5x}$ |
| $\cos(4x)$ or $\sin(4x)$ | $A\cos(4x) + B\sin(4x)$ |
| $e^{3x} \sin(4x)$ | $e^{3x}\bigl(A\cos(4x) + B\sin(4x)\bigr)$ |
| $5x^2 \sin(4x)$ | $(Ax^2 + Bx + C)\cos(4x) + (Ex^2 + Fx + G)\sin(4x)$ |
| $xe^{3x} \cos(4x)$ | $e^{3x}\bigl((Ax + B)\cos(4x) + (Cx + E)\sin(4x)\bigr)$ |
| $\cosh(2x)$ or $\sinh(2x)$ | $A\cosh(2x) + B\sinh(2x)$, or split into $e^{\pm 2x}$ |

**Form rule.** Without duplication, $y_p$ is a combination of all the independent functions  
obtained from the terms of $g$ by repeated differentiation.  
Use a separate set of coefficients for each term of $g$.

### Example 5: forms only

| Equation | Trial $y_p$ |
|---|---|
| $y'' - 8y' + 25y = 5x^3 e^{-x} - 7e^{-x}$ | $(Ax^3 + Bx^2 + Cx + E)e^{-x}$ |
| $y'' + 4y = x\cos x$ | $(Ax + B)\cos x + (Cx + E)\sin x$ |
| $y'' - 9y' + 14y = 3x^2 - 5\sin(2x) + 7xe^{6x}$ | $Ax^2 + Bx + C + E\cos(2x) + F\sin(2x) + (Gx + H)e^{6x}$ |

First check that no trial term solves the homogeneous equation:  
the auxiliary roots are $4 \pm 3i$, $\pm 2i$, and $2, 7$.

### The multiplication rule

**Multiplication rule.** If part of the trial duplicates a term of $y_c$,  
multiply that part by $x^k$, with $k$ the smallest power that removes all duplication.

Equivalently, for an input $x^d e^{mx}$, or $x^d e^{\alpha x}$ times $\cos(\beta x)$ or $\sin(\beta x)$,  
let $k$ be the multiplicity of $m$, or of $\alpha + i\beta$, as a root of $P$ ($k = 0$ if it is not a root).  
The trial is $x^k$ times the usual trial.

### Why the multiplication rule works

Suppose $m$ is a root of $P$ of multiplicity $k$.  
By Taylor's formula,
$$ P(D + m) = \sum_{j = 0}^n \frac{P^{(j)}(m)}{j!} D^j = \frac{P^{(k)}(m)}{k!} D^k + \cdots, $$
because $P(m) = P'(m) = \cdots = P^{(k - 1)}(m) = 0$.  
So $P(D)[e^{mx} f] = e^{mx} P(D + m) f$ differentiates $f$ at least $k$ times.  
To produce $x^d e^{mx}$, the polynomial $f$ must have degree $d + k$,  
and its terms of degree below $k$ only reproduce $y_c$; drop them.  
What remains is exactly $x^k(A_d x^d + \cdots + A_0)$.

For a pure exponential input this gives a formula, since $P(D + m)\,x^k = P^{(k)}(m)$:
$$ \boxed{P(D) y = e^{mx} \quad\Longrightarrow\quad y_p = \frac{x^k e^{mx}}{P^{(k)}(m)},} $$
where $k$ is the multiplicity of $m$; for $k = 0$ this is $e^{mx}/P(m)$.  
In Example 4, $P'(1) = 2 - 5 = -3$, so again $y_p = -\frac83 xe^x$.  
Use the formula to check answers; when a problem asks for undetermined coefficients, show the trial form.

### Example 6: a double root

For $y'' - 2y' + y = e^x$, $y_c = c_1 e^x + c_2 xe^x$, so both $Ae^x$ and $Axe^x$ duplicate.  
Here $P(m) = (m - 1)^2$, so $k = 2$ and the trial is $Ax^2 e^x$.  
The shift rule gives $P(D + 1) = D^2$ and $D^2(Ax^2) = 2A = 1$, so
$$ \boxed{y_p = \frac12 x^2 e^x.} $$
The formula agrees, since $P''(1) = 2$.

### Example 7: an initial value problem with duplication

Solve $y'' + y = 4x + 10\sin x$, $y(\pi) = 0$, $y'(\pi) = 2$.  
Here $y_c = c_1 \cos x + c_2 \sin x$, so $\sin x$ duplicates and the trial is
$$ y_p = Ax + B + x(C\cos x + E\sin x). $$
Using $(D^2 + 1)(x\cos x) = -2\sin x$ and $(D^2 + 1)(x\sin x) = 2\cos x$,
$$ Ax + B - 2C\sin x + 2E\cos x = 4x + 10\sin x, $$
so $A = 4$, $B = 0$, $C = -5$, $E = 0$, and $y_p = 4x - 5x\cos x$.

Apply the initial conditions to $y = c_1 \cos x + c_2 \sin x + 4x - 5x\cos x$:
$$ y(\pi) = -c_1 + 9\pi = 0, \qquad y'(\pi) = -c_2 + 9 = 2. $$
Hence
$$ \boxed{y = 9\pi\cos x + 7\sin x + 4x - 5x\cos x.} $$
Initial conditions go on the **general** solution $y_c + y_p$, never on $y_c$ alone.

### Example 8: two parts, one duplication

Solve $y'' - 6y' + 9y = 6x^2 + 2 - 12e^{3x}$, where $y_c = c_1 e^{3x} + c_2 xe^{3x}$.  
The polynomial part does not duplicate; $Ax^2 + Bx + C$ gives
$$ 9Ax^2 + (9B - 12A)x + (2A - 6B + 9C) = 6x^2 + 2, $$
so $A = \frac23$, $B = \frac89$, $C = \frac23$.  
The root $3$ is double, so the exponential part needs $Ex^2 e^{3x}$; $P(D + 3) = D^2$ gives $2E = -12$. Thus
$$ \boxed{y = c_1 e^{3x} + c_2 xe^{3x} + \frac23 x^2 + \frac89 x + \frac23 - 6x^2 e^{3x}.} $$

### Example 9: a higher-order form

For $y^{(4)} + y''' = 1 - x^2 e^{-x}$, the auxiliary polynomial is $m^3(m + 1)$, so  
$y_c = c_1 + c_2 x + c_3 x^2 + c_4 e^{-x}$.  
The usual trial $A + (Bx^2 + Cx + E)e^{-x}$ duplicates in both parts:  
$0$ is a triple root and $-1$ is a simple root. Therefore
$$ y_p = Ax^3 + (Bx^3 + Cx^2 + Ex)e^{-x}. $$

### Example 10: a $\cosh$ input

For $y'' + 3y' + 2y = \cosh x$, write $\cosh x = \frac12 e^x + \frac12 e^{-x}$.  
The roots are $-1$ and $-2$, so only the $e^{-x}$ part duplicates.  
With $P(1) = 6$ and $P'(-1) = 1$, the formula gives
$$ \boxed{y_p = \frac{e^x}{12} + \frac12 xe^{-x}.} $$
The trial $A\cosh x + B\sinh x$ would fail here; split $\cosh$ whenever $1$ or $-1$ is a root.

### Complex exponentials for sines and cosines

Since $P$ has real coefficients, if $z$ solves $P(D) z = e^{(\alpha + i\beta)x}$,  
then $\operatorname{Re} z$ solves $P(D) y = e^{\alpha x} \cos(\beta x)$  
and $\operatorname{Im} z$ solves $P(D) y = e^{\alpha x} \sin(\beta x)$.  
With $s = \alpha + i\beta$ of multiplicity $k$,
$$ \boxed{z_p = \frac{x^k e^{sx}}{P^{(k)}(s)}.} $$
One complex division replaces the $2 \times 2$ system for the cosine and sine coefficients.

### Example 11: Example 2 again, and a third-order equation

For $y'' - y' + y = 2\sin(3x)$, take $s = 3i$, so $P(3i) = -9 - 3i + 1 = -8 - 3i$ and
$$ y_p = \operatorname{Im}\frac{2e^{3ix}}{-8 - 3i} = \operatorname{Im}\frac{2(-8 + 3i)\bigl(\cos(3x) + i\sin(3x)\bigr)}{73}
= \frac{6\cos(3x) - 16\sin(3x)}{73}. $$
For $y''' + y'' = e^x \cos x$, take $s = 1 + i$, so $P(s) = s^2(s + 1) = 2i(2 + i) = -2 + 4i$ and
$$ y_p = \operatorname{Re}\frac{e^{(1 + i)x}}{-2 + 4i} = -\frac{1}{10}e^x \cos x + \frac15 e^x \sin x. $$

### Example 12: resonance

For $y'' + 4y = 8\sin(2x)$, the input frequency equals the natural frequency: $2i$ is a simple root.  
With $P'(s) = 2s$ and $P'(2i) = 4i$,
$$ y_p = \operatorname{Im}\frac{8xe^{2ix}}{4i} = \operatorname{Im}\bigl(-2ixe^{2ix}\bigr) = -2x\cos(2x). $$
The factor $x$ makes the amplitude grow without bound: this is **resonance**.

#### Picture: beats and resonance

Start from rest and drive $y'' + y = \cos(\omega t)$.  
For $\omega \ne 1$,
$$ y = \frac{\cos(\omega t) - \cos t}{1 - \omega^2} = \frac{2}{1 - \omega^2}\sin\frac{(1 - \omega)t}{2}\sin\frac{(1 + \omega)t}{2}, $$
and for $\omega = 1$ the multiplication rule gives $y = \frac12 t\sin t$.

```python
resonance_times = np.linspace(0, 80, 4001)
def driven_from_rest(omega, t):
    omega = float(omega)
    if omega == 1:
        return t*np.sin(t)/2
    return (np.cos(omega*t) - np.cos(t))/(1 - omega**2)
def response_plot(omega, envelope, color):
    result = line(list(zip(resonance_times, driven_from_rest(omega, resonance_times))), color=color,
                  legend_label=rf'$\omega = {float(omega)}$')
    for sign in [1, -1]:
        result += line(list(zip(resonance_times, sign*envelope)), color='gray', linestyle='--')
    return result
```

```python
beats_envelope = 2/(1 - 0.9**2)*np.abs(np.sin(0.05*resonance_times))
response_plot(0.9, beats_envelope, '#0072B2').show(xmin=0, xmax=80, ymin=-42, ymax=42, figsize=[9, 3.5],
    frame=True, axes=False, axes_labels=['$t$', '$y$'], legend_options={'loc': 'upper left'})
response_plot(1, resonance_times/2, '#D55E00').show(xmin=0, xmax=80, ymin=-42, ymax=42, figsize=[9, 3.5],
    frame=True, axes=False, axes_labels=['$t$', '$y$'], legend_options={'loc': 'upper left'})
```

**Read the picture:** at $\omega = 0.9$ the input and the natural oscillation drift in and out of phase.  
The amplitude swells to $2/(1 - \omega^2) \approx 10.5$ and collapses every $2\pi/(1 - \omega) \approx 63$ time units: **beats**.  
As $\omega \to 1$ the bursts grow taller and longer, and in the limit the envelope becomes the line $t/2$.  
The resonant solution $\frac12 t\sin t$ is the limit of beats, and the factor $t$ from the multiplication rule is that envelope.

## 4.5 Undetermined Coefficients: Annihilator Approach

### Factoring operators

A constant-coefficient operator factors like its auxiliary polynomial, and the factors commute:
$$ D^2 + 5D + 6 = (D + 2)(D + 3) = (D + 3)(D + 2). $$
This fails for variable coefficients: $D(xD) y = xy'' + y'$, while $(xD) D y = xy''$.

### Annihilators

An operator $L$ **annihilates** $f$ if $L f = 0$.  
By 4.3, the annihilators of the basic functions are read off from their roots:

| Functions | Root | Annihilator |
|---|---|---|
| $1, x, \ldots, x^{n - 1}$ | $0$, multiplicity $n$ | $D^n$ |
| $e^{\alpha x}, xe^{\alpha x}, \ldots, x^{n - 1} e^{\alpha x}$ | $\alpha$, multiplicity $n$ | $(D - \alpha)^n$ |
| $x^j e^{\alpha x} \cos(\beta x)$, $x^j e^{\alpha x} \sin(\beta x)$, $j < n$ | $\alpha \pm i\beta$, multiplicity $n$ | $\bigl(D^2 - 2\alpha D + \alpha^2 + \beta^2\bigr)^n$ |

For a sum, multiply the annihilators of the different root families,  
using the highest power needed in each family.  
The result is the annihilator of **least order**.  
For instance, $7 - x + 6\sin(3x)$ is annihilated by $D^2(D^2 + 9)$,  
and $\cosh x$ by $(D - 1)(D + 1) = D^2 - 1$.

### Example 1: annihilators of least order

| Function | Annihilator |
|---|---|
| $1 - 5x^2 + 8x^3$ | $D^4$ |
| $e^{-3x}$ | $D + 3$ |
| $4e^{2x} - 10xe^{2x}$ | $(D - 2)^2$ |
| $5e^{-x} \cos(2x) - 9e^{-x} \sin(2x)$ | $D^2 + 2D + 5$ |
| $x\sin(2x) + 3e^x$ | $(D^2 + 4)^2(D - 1)$ |

### Reading solutions from an operator

The question can be reversed.  
The functions annihilated by $D^3 + 6D^2 + 9D = D(D + 3)^2$ are the solutions of $(D^3 + 6D^2 + 9D) y = 0$:
$$ 1, \qquad e^{-3x}, \qquad xe^{-3x}. $$
Conversely, the general solution $c_1 e^{-x} + c_2 xe^{-x} + c_3 \cos(2x) + c_4 \sin(2x)$  
comes from $(D + 1)^2(D^2 + 4) y = 0$, that is,
$$ y^{(4)} + 2y''' + 5y'' + 8y' + 4y = 0. $$

### The method

To solve $L y = g$ with constant coefficients:

1. Solve $L y = 0$ for $y_c$.
2. Find an annihilator $L_1$ of $g$; every solution of $L y = g$ then solves $L_1 L y = 0$.
3. Solve the homogeneous equation $L_1 L y = 0$ by 4.3.
4. Delete the terms of $y_c$; the remaining terms form the trial $y_p$.
5. Substitute into $L y = g$ and match coefficients.

Roots shared by $L$ and $L_1$ become repeated roots of $L_1 L$: this is the multiplication rule.

### Example 2: $y'' + 3y' + 2y = 4x^2$

Here $L = (D + 1)(D + 2)$, and $D^3$ annihilates $4x^2$.  
Solutions of $D^3(D + 1)(D + 2) y = 0$ have the form
$$ c_1 e^{-x} + c_2 e^{-2x} + c_3 + c_4 x + c_5 x^2. $$
Removing $y_c$ leaves $y_p = Ax^2 + Bx + C$, and substitution gives
$$ 2Ax^2 + (6A + 2B)x + (2A + 3B + 2C) = 4x^2, $$
so $A = 2$, $B = -6$, $C = 7$, and $y = c_1 e^{-x} + c_2 e^{-2x} + 2x^2 - 6x + 7$.

### Example 3: $y'' - 3y' = 8e^{3x} + 4\sin x$

Here $L = D(D - 3)$, and $(D - 3)(D^2 + 1)$ annihilates the input.  
The combined operator $D(D - 3)^2(D^2 + 1)$ has the double root $3$, so
$$ y_p = Axe^{3x} + B\cos x + C\sin x. $$
The shift rule gives $L(Axe^{3x}) = e^{3x}(D + 3)D(Ax) = 3Ae^{3x}$, so $A = \frac83$.  
The trigonometric part gives $(-B - 3C)\cos x + (3B - C)\sin x = 4\sin x$, so $B = \frac65$ and $C = -\frac25$:
$$ \boxed{y = c_1 + c_2 e^{3x} + \frac83 xe^{3x} + \frac65 \cos x - \frac25 \sin x.} $$

### Example 4: $y'' + y = x\cos x - \cos x$

The input $(x - 1)\cos x$ is annihilated by $(D^2 + 1)^2$, and $L = D^2 + 1$.  
So $(D^2 + 1)^3$ appears, and the trial is
$$ y_p = (Ax^2 + Bx)\cos x + (Cx^2 + Ex)\sin x. $$
Matching coefficients gives
$$ \boxed{y_p = \frac14 x\cos x - \frac12 x\sin x + \frac14 x^2 \sin x.} $$

Complex exponentials shorten the computation.  
Write the input as $\operatorname{Re}\bigl[(x - 1)e^{ix}\bigr]$ and look for $z_p = e^{ix} f(x)$.  
Since $P(D + i) = (D + i)^2 + 1 = D^2 + 2iD$, the shift rule turns the equation into
$$ (D^2 + 2iD) f = x - 1. $$
With $f = ax^2 + bx$, this gives $a = -\frac i4$ and $b = \frac14 + \frac i2$,  
and $\operatorname{Re}\bigl[e^{ix} f\bigr]$ is the $y_p$ above.

### Example 5: forms only

| Equation | Annihilator of $g$ | Trial $y_p$ |
|---|---|---|
| $y'' - 2y' + y = 10e^{-2x} \cos x$ | $D^2 + 4D + 5$ | $e^{-2x}(A\cos x + B\sin x)$ |
| $y''' - 4y'' + 4y' = 5x^2 - 6x + 4x^2 e^{2x} + 3e^{5x}$ | $D^3(D - 2)^3(D - 5)$ | $Ax + Bx^2 + Cx^3 + (Ex^2 + Fx^3 + Gx^4)e^{2x} + He^{5x}$ |

In the second equation $L = D(D - 2)^2$, so the combined operator is $D^4(D - 2)^5(D - 5)$;  
the constant and the terms $e^{2x}$, $xe^{2x}$ belong to $y_c$ and are removed.

### Two descriptions of one method

| Superposition approach (4.4) | Annihilator approach (4.5) |
|---|---|
| derivatives of $g$ stay in a finite family | $g$ is annihilated by a constant-coefficient operator |
| table of trial forms | solutions of $L_1 y = 0$ |
| multiplication rule | shared roots of $L$ and $L_1$ become repeated roots |

Both lead to the same trial $y_p$ and the same equations for its coefficients.  
The common reason is the shift rule $P(D)[e^{mx} f] = e^{mx} P(D + m) f$.

### Summary

| Task | Tool |
|---|---|
| Solve $P(D) y = 0$ | roots of $P(m)$; extra factors $x^j$ for repeated roots; $e^{\alpha x}\cos$, $e^{\alpha x}\sin$ for complex pairs |
| Choose a trial $y_p$ | the table, or the solutions of $L_1 y = 0$; multiply by $x^k$ for a root of multiplicity $k$ |
| Check $y_p$ quickly | $y_p = x^k e^{sx}/P^{(k)}(s)$, with the real or imaginary part for cosines and sines |
| Solve an initial value problem | apply the conditions to $y_c + y_p$ |

### Practice habits for constant-coefficient equations

- Factor $P(m)$ completely before writing any trial.
- Write the roots of the input next to the roots of $P$; their overlaps decide the powers of $x$.
- Split $\cosh$ and $\sinh$ into exponentials when one of them duplicates.
- Keep a separate set of coefficients for each piece of the input.
- Check $y_p$ by substitution, or with the formula.

```python
from IPython.display import display, Math
R = PolynomialRing(QQ, 'm'); m = R.gen()
for P in [m**3 + 3*m**2 - 4, 3*m**3 + 5*m**2 + 10*m - 4, m**4 + 2*m**2 + 1, m**3 - 8]:
    display(Math(latex(P) + ' = ' + latex(P.factor())))
K = QuadraticField(2, 'sqrt2')
display(Math(latex(m**4 + 1) + ' = ' + latex((m**4 + 1).change_ring(K).factor())))
```

These are Examples 3, 5, and 4 of 4.3, and $m^3 - 8$ for $y''' = 8y$.  
Over $\mathbb Q$ the polynomial $m^4 + 1$ is irreducible;  
allowing $\sqrt2$ reproduces the factorization of Example 6.

```python
x = SR.var('x')
assume(x, 'real')
c1, c2, c3 = SR.var('c1 c2 c3')
def apply_P(coefficients, y):
    # coefficients = [a_0, a_1, ..., a_n] for a_n y^(n) + ... + a_1 y' + a_0 y
    return sum(a*diff(SR(y), x, k) for k, a in enumerate(coefficients))
```

```python
s_var = SR.var('s_var')
def response(coefficients, s):
    # y_p = x^k e^{sx} / P^(k)(s) for P(D) y = e^{sx}, where k is the multiplicity of s
    P = sum(a*s_var**k for k, a in enumerate(coefficients))
    k = 0
    while bool(P.diff(s_var, k).subs(s_var=s) == 0):
        k += 1
    return x**k*exp(s*x)/P.diff(s_var, k).subs(s_var=s)
```

## Bonus: Exponential Moving Averages

### Smoothing data one step at a time

A noisy sequence $x_1, x_2, \ldots$ is often smoothed by the **exponential moving average** (EMA)
$$ s_n = s_{n - 1} + \alpha(x_n - s_{n - 1}) = (1 - \alpha)s_{n - 1} + \alpha x_n, \qquad 0 < \alpha \le 1. $$
Each step moves the average a fraction $\alpha$ of the way toward the new data point.  
Unrolling the recursion,
$$ s_n = \alpha\sum_{j = 0}^{n - 1} (1 - \alpha)^j x_{n - j} + (1 - \alpha)^n s_0. $$
The newest point gets weight $\alpha$, and older points get weights shrinking by the factor $1 - \alpha$ per step.

EMAs appear in many places:

- a stock's $N$-day EMA uses $\alpha = 2/(N + 1)$;
- the momentum and Adam optimizers in machine learning keep EMAs of gradients, with $1 - \alpha = \beta = 0.9$;
- sensors and displays smooth their readings this way.

Starting from $s_0 = 0$, the weights sum to $1 - (1 - \alpha)^n$ instead of $1$;  
this is why Adam divides by $1 - \beta^n$, its bias correction.

### From the update rule to a differential equation

Let the data arrive every $\Delta t$ and choose $\alpha = \Delta t/\tau$ for a fixed **time constant** $\tau$.  
Then
$$ \frac{s_n - s_{n - 1}}{\Delta t} = \frac{x_n - s_{n - 1}}{\tau}, $$
and as $\Delta t \to 0$,
$$ \boxed{\tau s' + s = x(t).} $$
This is the RC circuit of 3.1, where the capacitor voltage averages the source with $\tau = RC$.  
It is also Newton's cooling with a changing room temperature,  
and the mixing tank whose concentration chases the incoming concentration.

### Solving with the shift rule

Write $s = e^{-t/\tau} f$.  
The shift rule gives $(D + 1/\tau)[e^{-t/\tau} f] = e^{-t/\tau} f'$, so the equation becomes
$$ f' = \frac1\tau e^{t/\tau} x(t). $$
Integrating from $0$ to $t$,
$$ \boxed{s(t) = s(0)e^{-t/\tau} + \frac1\tau\int_0^t e^{-(t - u)/\tau} x(u)\,du.} $$
The first term is the zero-input response of 4.1, and the second is the zero-state response.

The weights $\frac1\tau e^{-(t - u)/\tau}$ have total integral $1$ over the whole past,  
so $s(t)$ is an average of past inputs that forgets them exponentially.  
Factoring out $e^{-t/\tau}$ is the same step as multiplying by the integrating factor $e^{t/\tau}$ in 2.3.

### An EMA of an EMA

Smoothing twice with the same $\tau$ means $(D + 1/\tau) s_1 = x/\tau$ and $(D + 1/\tau) s_2 = s_1/\tau$, so
$$ \Bigl(D + \frac1\tau\Bigr)^2 s_2 = \frac{1}{\tau^2} x. $$
The auxiliary root $-1/\tau$ is now double.  
Writing $s_2 = e^{-t/\tau} f$, the shift rule gives $f'' = \frac{1}{\tau^2} e^{t/\tau} x$.  
Integrating twice from rest, with Cauchy's formula $\int_0^t \int_0^v h(u)\,du\,dv = \int_0^t (t - u) h(u)\,du$,
$$ s_2(t) = \frac{1}{\tau^2}\int_0^t (t - u) e^{-(t - u)/\tau} x(u)\,du. $$
The factor $t - u$ is the power of $x$ that a double root produces.

With $k$ identical stages, an input from lag $\ell = t - u$ ago gets the weight
$$ w_k(\ell) = \frac{\ell^{k - 1}}{(k - 1)!\,\tau^k} e^{-\ell/\tau}, $$
the solution pattern of a root of multiplicity $k$.  
In probability this is the Erlang density, the waiting time for $k$ independent exponential events.  
Each $w_k$ has total integral $1$, its peak at lag $(k - 1)\tau$, and its average lag $k\tau$.  
The critically damped oscillator $(D + 1)^2 y = g$ of 4.3 is two such stages with $\tau = 1$.

#### Picture: the weights of one, two, and three stages

```python
weight_plot = Graphics()
for k, color in [(1, '#0072B2'), (2, '#009E73'), (3, '#D55E00')]:
    weight_plot += plot(x**(k - 1)*exp(-x)/factorial(k - 1), (x, 0, 8), color=color, thickness=2,
                        legend_label=f'$k = {k}$')
weight_plot.show(xmin=0, xmax=8, figsize=[9, 3.5], frame=True, axes=False,
                 axes_labels=[r'lag $\ell/\tau$', r'weight $\tau w_k$'], legend_options={'loc': 'upper right'})
```

**Read the picture:** a single EMA gives the newest input the most weight.  
With two or three stages the newest input gets almost no weight, and the weights peak at lags $\tau$ and $2\tau$.  
Every curve has area $1$, so each cascade is still an average:  
the extra stages trade delay for smoothness.

#### Simulation: smoothing a noisy step

Feed a noisy step, which jumps from $0$ to $1$ at step $40$, into one, two, and three EMA stages with $\alpha = 0.1$.  
In steps, the time constant is $\tau = 1/\alpha = 10$.

```python
rng = np.random.default_rng(int(1021))
ema_signal = np.where(np.arange(120) < 40, 0.0, 1.0) + float(0.3)*rng.standard_normal(120)
def ema(data, alpha=float(0.1)):
    s, out = 0.0, []
    for value in data:
        s += alpha*(float(value) - s)
        out.append(s)
    return np.array(out)
ema_stages = [ema_signal]
for _ in range(3):
    ema_stages.append(ema(ema_stages[-1]))
```

```python
ema_plot = points(list(enumerate(ema_signal)), color='silver', size=12, legend_label='noisy data')
for k, color in [(1, '#0072B2'), (2, '#009E73'), (3, '#D55E00')]:
    ema_plot += line(list(enumerate(ema_stages[k])), color=color, thickness=2,
                     legend_label=f'{k} stage' + ('s' if k > 1 else ''))
ema_plot.show(xmin=0, xmax=120, figsize=[9, 4], frame=True, axes=False,
              axes_labels=['step $n$', 'value'], legend_options={'loc': 'upper left'})
```

**Read the picture:** one stage reacts first but keeps much of the noise.  
Each extra stage removes more noise and adds about $\tau = 10$ more steps of delay, since the average lag is $k\tau$.  
The two- and three-stage curves leave the old level smoothly instead of with a kink,  
because their weights vanish at lag $0$.

### The discrete picture

The recursion itself is a first-order difference equation with characteristic root $r = 1 - \alpha$:  
$s_n = Cr^n$ solves the homogeneous recursion, just as $e^{mt}$ solves $s' = ms$,  
and $r = 1 - \Delta t/\tau \approx e^{-\Delta t/\tau}$.  
Two EMA stages give the double root $r$, and the extra solution is $nr^n$,  
the discrete twin of $te^{mt}$.
