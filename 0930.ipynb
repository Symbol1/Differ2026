## 2.4 Exact Equations

### The differential of a function of two variables

If a function $f(x, y)$ has continuous first partial derivatives,  
its **differential** is defined as
$$ df = \frac{\partial f}{\partial x}\,dx + \frac{\partial f}{\partial y}\,dy. $$
You can think of $df$ as a function from $\mathbb{R}^2 \to \mathbb{R}^2$,  
except that the first output is appended by $dx$  
(something happening in $x$-direction)  
and the second output is appended by $dy$.

In general, a differential $1$-form is $g(x, y)\,dx + h(x, y)\,dy$.  
It can be used to detect if an object is moving in $x$-direction,  
if so, multiply that by $g(x, y)$ and integrate.

Along a level curve $f(x, y) = c$, the value of $f$ does not change,  
so the (differential) equation that defines the level curve is
$$ \frac{\partial f}{\partial x}\,dx + \frac{\partial f}{\partial y}\,dy = 0. $$
The family of level curves share the same first-order differential equation.

For example, the family $x^2 - 5xy + y^3 = c$ gives
$$ (2x - 5y)\,dx + (3y^2 - 5x)\,dy = 0, $$
or, where $3y^2 - 5x \ne 0$,
$$ \frac{dy}{dx} = \frac{5y - 2x}{3y^2 - 5x}. $$
An exact equation reverses this process:  
start from the differential equation and find the function $f$.

### Exact equations

An expression $M(x, y)\,dx + N(x, y)\,dy$ is an **exact differential** in a region $R$  
if there is some function $f$ defined in $R$ such that $df$ recovers the given differential, that is,
$$ \frac{\partial f}{\partial x} = M, \qquad \frac{\partial f}{\partial y} = N. $$
The equation
$$ M(x, y)\,dx + N(x, y)\,dy = 0 $$
is an **exact equation** if its left-hand side is an exact differential.  
The solutions of the exact equation are exactly the level curves
$$ \boxed{f(x, y) = c.} $$

![alt text](0930/image.png)

For example,
$$ x^2 y^3\,dx + x^3 y^2\,dy = 0 $$
is exact, because
$$ d\left(\frac{x^3 y^3}{3}\right) = x^2 y^3\,dx + x^3 y^2\,dy. $$
So its solutions satisfy $x^3 y^3 = c$, or $xy = c$ after renaming the constant.

### Criterion for an exact differential

Let $M$ and $N$ be continuous and have continuous first partial derivatives  
on an open rectangle $R$.  
Then $M\,dx + N\,dy$ is an exact differential in $R$ if and only if
$$ \boxed{\frac{\partial M}{\partial y} = \frac{\partial N}{\partial x}.} $$
With subscripts for partial derivatives, the condition reads $M_y = N_x$.

**Proof of Necessity.**  
If $M = f_x$ and $N = f_y$, then
$$ M_y = f_{xy} = f_{yx} = N_x. $$
The mixed partial derivatives agree because they are continuous (Clairaut's theorem).

**Proof of Sufficiency.**  
The method below constructs $f$ whenever $M_y = N_x$ on $R$.

**Remark.**  
The criterion is a **zero-curl** condition.  
Regard $(M, N)$ as the vector field $(M, N, 0)$ in space; then
$$ \nabla \times (M, N, 0) = \left(0, 0, \frac{\partial N}{\partial x} - \frac{\partial M}{\partial y}\right). $$
An exact differential corresponds to a gradient field $(M, N) = \nabla f$,  
and the curl of a gradient is always zero.  

curl = 0 does not necessarily imply it's a gradient.  
But a rectangle it does imply.
So $(M, N)$ is a gradient field exactly when it is curl-free.

### Method of solution: a line integral

A **path** in $R$ is a piecewise smooth curve
$$ \gamma(\tau) = \bigl(x(\tau), y(\tau)\bigr), \qquad a \le \tau \le b. $$
Along $\gamma$, the 1-form detects motion in each direction and weights it:
$$ \int_\gamma M\,dx + N\,dy = \int_\gamma (M, N) \cdot d\gamma = \int_a^b \bigl[M(\gamma(\tau)) x'(\tau) + N(\gamma(\tau)) y'(\tau)\bigr]\,d\tau. $$
Motion in the $x$-direction is weighted by $M$, and motion in the $y$-direction by $N$.

Fix a base point $(x_0, y_0)$ in $R$.  
For each $(x, y)$ in $R$, choose a path $\gamma$ in $R$ from $(x_0, y_0)$ to $(x, y)$, and set
$$ \boxed{f(x, y) = \int_\gamma M\,dx + N\,dy.} $$
This defines a function only if the value does not depend on the choice of $\gamma$.  
We prove this path independence first, and then show that $df = M\,dx + N\,dy$.

#### Path independence

**Theorem.**  
If $M_y = N_x$ on the rectangle $R$,  
then any two paths in $R$ with the same endpoints give the same value of $\int_\gamma M\,dx + N\,dy$.

For example, go from $(0, 0)$ to $(1, 1)$ along $y = x$, or along the $x$-axis and then up.  
The exact form $x^2 y^3\,dx + x^3 y^2\,dy$ gives $1/3$ on both paths.  
The form $-y\,dx + x\,dy$, whose curl is $N_x - M_y = 2$, gives $0$ and $1$.

![alt text](0930/image-1.png)

**Proof.**  
Let $\gamma_0$ and $\gamma_1$ be paths in $R$ from $P$ to $Q$, both parametrized by $a \le \tau \le b$ (reparametrize if necessary).  
Slide $\gamma_0$ onto $\gamma_1$ through the family
$$ \gamma_\lambda(\tau) = (1 - \lambda)\gamma_0(\tau) + \lambda\gamma_1(\tau), \qquad 0 \le \lambda \le 1. $$
Each $\gamma_\lambda$ runs from $P$ to $Q$, and it stays in $R$ because a rectangle is convex.  
Write $\gamma_\lambda(\tau) = \bigl(x(\lambda, \tau), y(\lambda, \tau)\bigr)$ and $I(\lambda) = \int_{\gamma_\lambda} M\,dx + N\,dy$.  
It suffices to show that $I'(\lambda) = 0$.

Use subscripts $\lambda$ and $\tau$ for partial derivatives,  
and evaluate $M$, $N$, and their derivatives at $\gamma_\lambda(\tau)$.  
The chain rule gives $M_\lambda = M_x x_\lambda + M_y y_\lambda$ and $M_\tau = M_x x_\tau + M_y y_\tau$,  
and similarly for $N$.  
Expanding both sides, the terms with $x_{\lambda\tau}$, $y_{\lambda\tau}$, $M_x$, and $N_y$ cancel, leaving
$$ \frac{\partial}{\partial \lambda}(Mx_\tau + Ny_\tau) = \frac{\partial}{\partial \tau}(Mx_\lambda + Ny_\lambda) + (N_x - M_y)(x_\lambda y_\tau - x_\tau y_\lambda). $$

Differentiate under the integral sign and integrate this identity over $a \le \tau \le b$:
$$ I'(\lambda) = \Bigl[Mx_\lambda + Ny_\lambda\Bigr]_{\tau = a}^{\tau = b} + \int_a^b (N_x - M_y)(x_\lambda y_\tau - x_\tau y_\lambda)\,d\tau. $$
The endpoints $P$ and $Q$ do not move, so $x_\lambda = y_\lambda = 0$ at $\tau = a$ and $\tau = b$;  
the bracket vanishes.  
The zero-curl condition $N_x - M_y = 0$ kills the integral.  
So $I$ is constant, and the two paths give $I(0) = I(1)$.  
For piecewise smooth paths, apply this on each smooth piece; the brackets at the joints cancel. $\blacksquare$

**Geometric meaning.**  
The factor $x_\lambda y_\tau - x_\tau y_\lambda$ is the Jacobian of $(\lambda, \tau) \mapsto \gamma_\lambda(\tau)$,  
the rate at which the sliding path sweeps out signed area.  
So $I(1) - I(0)$ is the curl integrated over the region between the two paths: this is Green's theorem.  
In the example above, curl $2$ times area $1/2$ accounts for the difference $1 - 0$.  
A curl-free field lets the path slide freely; a hole in the region would block the slide.

#### The line integral is a potential

Path independence lets us choose convenient paths.  
To reach $(x + h, y)$, go to $(x, y)$ first and then along the horizontal segment, where $dy = 0$:
$$ f(x + h, y) - f(x, y) = \int_x^{x + h} M(t, y)\,dt. $$
Dividing by $h$ and letting $h \to 0$ gives $f_x = M$ by the fundamental theorem of calculus.  
A final vertical segment gives $f_y = N$ in the same way.  
Hence $df = M\,dx + N\,dy$: the solutions are exactly the level curves $f(x, y) = c$,  
and the solution through $(x_0, y_0)$ is $f(x, y) = 0$.

#### Computing $f$: choose a path

Any path gives the same $f$, so choose one that makes the integral easy.  
On a horizontal segment $dy = 0$ and only $M$ contributes; on a vertical segment only $N$ does.  
Going vertically from $(x_0, y_0)$ and then horizontally,
$$ f(x, y) = \int_{y_0}^{y} N(x_0, s)\,ds + \int_{x_0}^{x} M(t, y)\,dt. $$
Going horizontally first,
$$ f(x, y) = \int_{x_0}^{x} M(t, y_0)\,dt + \int_{y_0}^{y} N(x, s)\,ds. $$

#### The textbook summary

1. Check that $M_y = N_x$.
2. Integrate $M$ with respect to $x$, holding $y$ fixed: $f(x, y) = \int M\,dx + g(y)$.
3. Differentiate in $y$ and require $f_y = N$: $g'(y) = N - \dfrac{\partial}{\partial y}\displaystyle\int M\,dx$.
4. Integrate $g'$ to find $g$; the solutions are $f(x, y) = c$.

Examples 1–3 below follow these steps.

**The textbook way is a consequence of line integration.**  
Take the vertical-then-horizontal path from $(x_0, y_0)$:
$$ f(x, y) = \underbrace{\int_{y_0}^{y} N(x_0, s)\,ds}_{g(y)} + \int_{x_0}^{x} M(t, y)\,dt. $$
The second term is an antiderivative of $M$ in $x$, and the first term depends on $y$ alone.  
So Step 2 is this line integral,  
and the unknown $g(y)$ is the integral along the vertical segment.

Step 3 recovers that vertical segment.  
Differentiating under the integral sign and using $M_y = N_x$,
$$ N(x, y) - \frac{\partial}{\partial y}\int_{x_0}^{x} M(t, y)\,dt = N(x, y) - \int_{x_0}^{x} N_x(t, y)\,dt = N(x_0, y), $$
which is exactly $g'(y)$.  
This is why the textbook's $g'(y)$ never depends on $x$:  
it is $N$ read along the line $x = x_0$.

An indefinite $\int M\,dx$ differs from $\int_{x_0}^{x} M(t, y)\,dt$ by a function of $y$ alone,  
which $g$ absorbs;  
the constant left after integrating $g'$ only shifts the level $c$.  
Starting from $f = \int N\,dy + h(x)$, as in Examples 2 and 3, is the horizontal-then-vertical path:  
there $h(x) = \int_{x_0}^{x} M(t, y_0)\,dt$ is the integral along the horizontal segment.

### Example 1: $2xy\,dx + (x^2 - 1)\,dy = 0$

Here $M = 2xy$ and $N = x^2 - 1$, so
$$ \frac{\partial M}{\partial y} = 2x = \frac{\partial N}{\partial x}. $$
Integrating $f_x = 2xy$ with respect to $x$ gives
$$ f(x, y) = x^2 y + g(y). $$
Then $f_y = x^2 + g'(y) = x^2 - 1$, so $g'(y) = -1$ and $g(y) = -y$.

The solutions are
$$ \boxed{x^2 y - y = c,} $$
or explicitly
$$ y = \frac{c}{x^2 - 1} $$
on any interval not containing $x = \pm1$.  
The left side of the equation is the differential $d\bigl[(x^2 - 1)y\bigr]$,  
so this linear equation was exact from the start.

### Example 2: integrate $N$ first

Solve
$$ \bigl(e^{2y} - y\cos(xy)\bigr)\,dx + \bigl(2xe^{2y} - x\cos(xy) + 2y\bigr)\,dy = 0. $$
The equation is exact:
$$ \frac{\partial M}{\partial y} = 2e^{2y} + xy\sin(xy) - \cos(xy) = \frac{\partial N}{\partial x}. $$
Every term of $N$ is easy to integrate with respect to $y$.

Start with $f_y = N$:
$$ f(x, y) = xe^{2y} - \sin(xy) + y^2 + h(x). $$
Then
$$ f_x = e^{2y} - y\cos(xy) + h'(x) = M, $$
so $h'(x) = 0$ and $h$ is constant.  
The solutions are
$$ \boxed{xe^{2y} - \sin(xy) + y^2 = c.} $$

### Example 3: an initial value problem

Solve
$$ \frac{dy}{dx} = \frac{xy^2 - \cos(x)\sin(x)}{y(1 - x^2)}, \qquad y(0) = 2. $$
First write the equation in the form $M\,dx + N\,dy = 0$:
$$ \bigl(\cos(x)\sin(x) - xy^2\bigr)\,dx + y(1 - x^2)\,dy = 0. $$
It is exact, since
$$ \frac{\partial M}{\partial y} = -2xy = \frac{\partial N}{\partial x}. $$

Start with $f_y = N$:
$$ f(x, y) = \frac{y^2}{2}(1 - x^2) + h(x). $$
Then $f_x = -xy^2 + h'(x) = \cos(x)\sin(x) - xy^2$, so
$$ h'(x) = \cos(x)\sin(x), \qquad h(x) = -\frac{\cos(x)^2}{2}. $$
Multiply by $2$ and rename the constant:
$$ y^2(1 - x^2) - \cos(x)^2 = c. $$

The initial condition gives $c = 4 - 1 = 3$, so
$$ \boxed{y^2(1 - x^2) - \cos(x)^2 = 3.} $$
Since $y(0) = 2 > 0$, take the positive branch:
$$ \boxed{y = \sqrt{\frac{3 + \cos(x)^2}{1 - x^2}}, \qquad -1 < x < 1.} $$
The solution blows up as $x \to \pm1$, where the coefficient $1 - x^2$ vanishes.

The line integral from the initial point $(x_0, y_0) = (0, 2)$ reaches the same curve without solving for $c$.  
Along the vertical-then-horizontal path, $N(0, s) = s$ and
$$ f(x, y) = \int_2^y s\,ds + \int_0^x \bigl(\cos(t)\sin(t) - ty^2\bigr)\,dt = \frac{y^2 - 4}{2} + \frac{\sin(x)^2}{2} - \frac{x^2 y^2}{2}. $$
Setting $f = 0$ and using $\sin(x)^2 = 1 - \cos(x)^2$ gives $y^2(1 - x^2) - \cos(x)^2 = 3$ again.

### Integrating factors

If $M\,dx + N\,dy = 0$ is not exact, multiply it by a function $\mu(x, y)$  
chosen so that $\mu M\,dx + \mu N\,dy = 0$ is exact:
$$ (\mu M)_y = (\mu N)_x \quad\Longleftrightarrow\quad \mu_x N - \mu_y M = (M_y - N_x)\mu. $$
Such a $\mu$ is an **integrating factor**.  
The condition is a partial differential equation for $\mu$,  
usually harder than the original problem.  

So we go for a special case where $\mu$ depending on only one variable.

If $\mu = \mu(x)$, then $\mu_y = 0$ and
$$ \frac{d\mu}{dx} = \frac{M_y - N_x}{N} \mu. $$
This works when $(M_y - N_x)/N$ depends on $x$ alone,  
in which case we already know how to solve
$$ \boxed{\mu(x) = \exp\left(\int \frac{M_y - N_x}{N}\,dx\right).} $$
Similarly, if $(N_x - M_y)/M$ depends on $y$ alone,  
we choose
$$ \boxed{\mu(y) = \exp\left(\int \frac{N_x - M_y}{M}\,dy\right).} $$

### Example 4: a nonexact equation made exact

Solve
$$ xy\,dx + (2x^2 + 3y^2 - 20)\,dy = 0. $$
Here $M_y = x$ and $N_x = 4x$, so the equation is not exact.  
The first ratio depends on both variables:
$$ \frac{M_y - N_x}{N} = \frac{-3x}{2x^2 + 3y^2 - 20}. $$
The second depends on $y$ alone:
$$ \frac{N_x - M_y}{M} = \frac{3x}{xy} = \frac3y. $$

Hence $\mu = e^{3\ln|y|} = |y|^3$, and we may use $\mu = y^3$.  
The new equation
$$ xy^4\,dx + (2x^2 y^3 + 3y^5 - 20y^3)\,dy = 0 $$
is exact: both partial derivatives equal $4xy^3$.  
The method of solution gives
$$ \boxed{\frac12 x^2 y^4 + \frac12 y^6 - 5y^4 = c.} $$

Multiplying by $\mu$ may gain or lose solutions where $\mu$ vanishes or is undefined.  
Here $\mu = 0$ on the line $y = 0$, which does solve the original equation.  
It appears at the level $c = 0$, since
$$ \frac12 x^2 y^4 + \frac12 y^6 - 5y^4 = \frac{y^4}{2}(x^2 + y^2 - 10). $$
The same level also contains the circle $x^2 + y^2 = 10$.

### Integrating factors are not unique

The equation $x\,dy - y\,dx = 0$ is not exact: $M = -y$ and $N = x$ give $M_y = -1 \ne 1 = N_x$.  
Each factor below makes it exact.

| $\mu$ | $\mu (x\,dy - y\,dx)$ |
|---|---|
| $1/x^2$ | $d(y/x)$ |
| $1/y^2$ | $d(-x/y)$ |
| $1/(xy)$ | $d\ln\lvert y/x\rvert$ |
| $1/(x^2 + y^2)$ | $d\arctan(y/x)$ |

![alt text](0930/image-2.png)

Each potential is constant exactly along lines through the origin.  
So all four factors give the same solution curves,  
away from the points where the factor is undefined.  
Different integrating factors lead to different potentials, not different solutions.

### Earlier methods as integrating factors

**Linear equations.**  
Write $y' + P(x)y = f(x)$ as
$$ \bigl(P(x)y - f(x)\bigr)\,dx + dy = 0. $$
Then $(M_y - N_x)/N = P(x)$ depends on $x$ alone, and
$$ \mu(x) = e^{\int P(x)\,dx}, $$
exactly the integrating factor of 2.3.

**Separable equations.**  
Write $y' = g(x)h(y)$ as $g(x)h(y)\,dx - dy = 0$.  
This is usually not exact, but $\mu = 1/h(y)$ gives
$$ g(x)\,dx - \frac{dy}{h(y)} = 0, $$
which is exact because $M_y = 0 = N_x$.  
Separating variables is multiplying by an integrating factor;  
this is why dividing by $h(y)$ can lose the equilibrium solutions.

## 2.5 Solutions by Substitutions

### Change the unknown function

A substitution replaces the unknown $y$ by a new unknown $u$,  
chosen so that the new equation is separable or linear.  
If $y = g(x, u)$, the chain rule gives
$$ \frac{dy}{dx} = \frac{\partial g}{\partial x} + \frac{\partial g}{\partial u}\frac{du}{dx}. $$
Substitute into $y' = f(x, y)$, solve the new equation for $u$, and transform back to $y$.

Two examples:

- the Gompertz equation became linear after $u = \ln P$;
- exchanging the roles of $x$ and $y$ made $y' = 1/(x + y^2)$ linear in $x(y)$.

This section collects substitutions that work for whole classes of equations.

### Homogeneous equations

A function $f(x, y)$ is **homogeneous of degree $\alpha$** if
$$ f(tx, ty) = t^\alpha f(x, y) \qquad \text{for all } t > 0. $$
For example, $x^3 + y^3$ is homogeneous of degree $3$,  
while $x^3 + y^3 + 1$ is not homogeneous.

The equation $M\,dx + N\,dy = 0$ is **homogeneous**  
if $M$ and $N$ are homogeneous of the same degree.  
This meaning is unrelated to a homogeneous linear equation $y' + P(x)y = 0$.

Let $u = y/x$.  
For $x > 0$, take $t = x$ in the definition:
$$ M(x, y) = M(x \cdot 1, x \cdot u) = x^\alpha M(1, u), \qquad N(x, y) = x^\alpha N(1, u). $$
The factor $x^\alpha$ cancels in the slope, so
$$ \frac{dy}{dx} = -\frac{M(1, u)}{N(1, u)} = F\left(\frac yx\right). $$
For $x < 0$, take $t = -x$; the slope is again a function of $y/x$.

### The substitution $y = ux$

Put $y = ux$, where $u$ is a new unknown function of $x$.  
Then
$$ \frac{dy}{dx} = u + x\frac{du}{dx}, $$
and $y' = F(y/x)$ becomes
$$ x\frac{du}{dx} = F(u) - u. $$
This is separable:
$$ \boxed{\frac{du}{F(u) - u} = \frac{dx}{x}.} $$
Each root $u_0$ of $F(u) = u$ gives a line $y = u_0 x$ through the origin; check these separately.

### Example 1: $(x^2 + y^2)\,dx + (x^2 - xy)\,dy = 0$

Both coefficients are homogeneous of degree $2$.  
Substitute $y = ux$ and $dy = u\,dx + x\,du$:
$$ (x^2 + u^2 x^2)\,dx + (x^2 - ux^2)(u\,dx + x\,du) = 0. $$
Collecting terms,
$$ x^2(1 + u)\,dx + x^3(1 - u)\,du = 0. $$
Divide by $x^3(1 + u)$ and use $(1 - u)/(1 + u) = -1 + 2/(1 + u)$:
$$ \left(-1 + \frac{2}{1 + u}\right)du + \frac{dx}{x} = 0. $$

Integrating,
$$ -u + 2\ln|1 + u| + \ln|x| = \ln|c|. $$
Return to $u = y/x$ and combine the logarithms:
$$ \ln\frac{(x + y)^2}{|cx|} = \frac yx. $$
Exponentiate and absorb the sign into $c$:
$$ \boxed{(x + y)^2 = cxe^{y/x}.} $$
Dividing by $1 + u$ lost the line $y = -x$; it is included as $c = 0$.

### Choosing between $y = ux$ and $x = vy$

The substitution $x = vy$, with $dx = v\,dy + y\,dv$, also works for every homogeneous equation.  
Since $dx$ multiplies $M$, try $x = vy$ when $M$ is simpler than $N$.  
If one substitution leads to a hard integral, try the other.

For example, in $y\,dx - 2(x + y)\,dy = 0$ the coefficient $M = y$ is simple.  
With $x = vy$,
$$ y(v\,dy + y\,dv) - 2(vy + y)\,dy = 0 \quad\Longrightarrow\quad y^2\,dv = (v + 2)y\,dy. $$
Separating gives $\ln|v + 2| = \ln|y| + c_1$, so $v + 2 = cy$ and
$$ \boxed{x = cy^2 - 2y.} $$
The substitution $y = ux$ reaches the same answer through a longer partial-fraction calculation.  
The line $y = 0$ also satisfies the equation and must be added separately.

### Bernoulli's equation

A **Bernoulli equation** has the form
$$ \frac{dy}{dx} + P(x)y = f(x)y^n, $$
where $n$ is a real number.  
For $n = 0$ or $n = 1$ it is linear.  
For every other $n$, the substitution
$$ \boxed{u = y^{1 - n}} $$
turns it into a linear equation.

Multiply the equation by $(1 - n)y^{-n}$ and use $u' = (1 - n)y^{-n} y'$:
$$ \boxed{\frac{du}{dx} + (1 - n)P(x)u = (1 - n)f(x).} $$
Solve this with an integrating factor, then return to $y$.  
For $n > 0$, the function $y = 0$ is also a solution; check whether the final formula includes it.

For example, $y' = x\sqrt y$ from 2.2 has $n = 1/2$, and $u = \sqrt y$ satisfies $u' = x/2$.

### Example 2: $xy' + y = x^2 y^2$

Divide by $x$:
$$ y' + \frac1x y = xy^2. $$
This is a Bernoulli equation with $n = 2$.  
Put $u = y^{-1}$; then $y = u^{-1}$ and $y' = -u^{-2} u'$, so
$$ -u^{-2} u' + \frac1x u^{-1} = xu^{-2}. $$
Multiply by $-u^2$:
$$ u' - \frac1x u = -x. $$

On $x > 0$, the integrating factor is $e^{-\ln x} = x^{-1}$:
$$ \frac{d}{dx}\bigl(x^{-1} u\bigr) = -1, \qquad x^{-1} u = -x + c. $$
Hence $u = cx - x^2$ and
$$ \boxed{y = \frac{1}{cx - x^2}.} $$
The solution $y = 0$ is not in this family; it is a singular solution.  
Use an interval that avoids $x = 0$ and $x = c$.

### Another route: spot a product derivative

The left side of Example 2 is a product derivative: $xy' + y = (xy)'$.  
With $w = xy$, the equation becomes
$$ w' = w^2, $$
so $w = -1/(x + C)$ and
$$ y = \frac wx = -\frac{1}{x^2 + Cx}. $$
This is the same family with $c = -C$, and $y = 0$ is simply the equilibrium $w = 0$.  
Look for combinations such as $xy$, $y/x$, or $x + y$ that appear repeatedly.

### The logistic equation is a Bernoulli equation

The logistic equation
$$ P' = aP - bP^2 $$
is a Bernoulli equation with $n = 2$.  
With $u = 1/P$,
$$ u' = -\frac{P'}{P^2} = -\frac aP + b = -au + b. $$
This is linear, with equilibrium $u = b/a = 1/K$.

Its solution with $P(0) = P_0 > 0$ is
$$ u = \frac ba + \left(\frac1{P_0} - \frac ba\right)e^{-at}, $$
so
$$ P(t) = \frac1u = \frac{aP_0}{bP_0 + (a - bP_0)e^{-at}}. $$
The same formula will appear again in section 3.2.

### Reduction to separation of variables

An equation of the form
$$ \frac{dy}{dx} = f(Ax + By + C), \qquad B \ne 0, $$
becomes separable with
$$ \boxed{u = Ax + By + C.} $$
Indeed, $u' = A + By' = A + Bf(u)$, so
$$ \frac{du}{A + Bf(u)} = dx. $$
Each root $u_0$ of $A + Bf(u) = 0$ gives a line $Ax + By + C = u_0$ that solves the equation.

### Example 3: $y' = (-2x + y)^2 - 7$, $y(0) = 0$

Let $u = -2x + y$.  
Then $u' = -2 + y'$, so
$$ u' + 2 = u^2 - 7, \qquad u' = u^2 - 9. $$
This has the same form as $y' = y^2 - 4$ in 2.2.  
Partial fractions give
$$ \frac16\ln\left|\frac{u - 3}{u + 3}\right| = x + c_1, \qquad \frac{u - 3}{u + 3} = ce^{6x}. $$

Solve for $u$ and use $y = 2x + u$:
$$ y = 2x + 3\frac{1 + ce^{6x}}{1 - ce^{6x}}. $$
The condition $y(0) = 0$ means $u(0) = 0$, so $c = -1$:
$$ \boxed{y = 2x + 3\frac{1 - e^{6x}}{1 + e^{6x}} = 2x - 3\tanh(3x).} $$
This solution is defined for every real $x$.

### Example 4: $xy' = y\ln(xy)$

The combination $xy$ appears inside the logarithm, so try $u = xy$, with $u > 0$.  
Then $u' = y + xy' = y + y\ln u$, and $y = u/x$ gives
$$ \frac{du}{dx} = \frac{u(1 + \ln u)}{x}. $$
Separate the variables; the substitution $w = 1 + \ln u$ evaluates the integral:
$$ \int\frac{du}{u(1 + \ln u)} = \ln|1 + \ln u| = \ln|x| + c_1. $$

Hence $1 + \ln u = cx$, $u = e^{cx - 1}$, and
$$ \boxed{y = \frac{e^{cx - 1}}{x}.} $$
The value $c = 0$ gives $xy = e^{-1}$,  
the solution lost when dividing by $1 + \ln u$.

### Riccati's equation

A **Riccati equation** has the form
$$ \frac{dy}{dx} = P(x) + Q(x)y + R(x)y^2. $$
If one solution $y_1$ is known, put $y = y_1 + u$.  
The terms without $u$ cancel because $y_1$ is a solution, leaving
$$ \frac{du}{dx} = (Q + 2Ry_1)u + Ru^2. $$
This is a Bernoulli equation with $n = 2$, so $w = 1/u$ makes it linear.

### Example 5: a Riccati equation with a known solution

Solve
$$ \frac{dy}{dx} = -\frac{4}{x^2} - \frac yx + y^2, \qquad x > 0, $$
using the known solution $y_1 = 2/x$.  
Here $Q + 2Ry_1 = -1/x + 4/x = 3/x$, so $y = 2/x + u$ gives
$$ u' = \frac3x u + u^2. $$

Put $w = 1/u$.  
Then $w' = -u'/u^2$, and
$$ w' + \frac3x w = -1. $$
The integrating factor $x^3$ gives $(x^3 w)' = -x^3$, so
$$ w = -\frac x4 + \frac{C}{x^3}, \qquad u = \frac1w = \frac{4x^3}{c - x^4}, \quad c = 4C. $$
Therefore,
$$ \boxed{y = \frac2x + \frac{4x^3}{c - x^4}.} $$

The substitution $u = xy$ from Example 4 also works here:  
it gives the separable equation $xu' = u^2 - 4$,  
whose equilibria $u = \pm2$ are the solutions $y = \pm2/x$.  
With three known solutions, the cross-ratio formula from the 0916 homework  
produces every other solution without any integration.

### Summary: choosing a substitution

| Form of the equation | Substitution | New equation |
|---|---|---|
| $y' = F(y/x)$: $M$, $N$ homogeneous of the same degree | $y = ux$ or $x = vy$ | separable |
| $y' + P(x)y = f(x)y^n$, $n \ne 0, 1$ | $u = y^{1 - n}$ | linear |
| $y' = f(Ax + By + C)$, $B \ne 0$ | $u = Ax + By + C$ | separable |
| $y' = P + Qy + Ry^2$ with a known solution $y_1$ | $y = y_1 + u$, then $w = 1/u$ | linear |

After any substitution, transform back, look for lost solutions,  
and verify in the original equation.

### Practice habits for exact equations and substitutions

- Put the equation in the form $M\,dx + N\,dy = 0$ before testing $M_y = N_x$.
- Find $g(y)$ from $f_y = N$; adding $\int M\,dx$ and $\int N\,dy$ usually counts some terms twice.
- After multiplying by $\mu$ or dividing by an expression, check for solutions gained or lost.
- Look for combinations such as $y/x$, $xy$, or $Ax + By$ that appear repeatedly.
- Verify the answer by differentiating, implicitly if necessary.

## Checking Worked Solutions with Sage

Sage can test exactness, construct a potential, and verify a proposed solution.  
A symbolic check confirms the algebra;  
domains, branches, and lost solutions still need a separate look.

```python
def potential(M, N):
    assert (M.diff(y) - N.diff(x)).simplify_full() == 0, 'not exact'
    f = M.integrate(x)
    g_prime = (N - f.diff(y)).simplify_full()
    assert g_prime.diff(x).simplify_full() == 0
    return f + g_prime.integrate(y)
```
