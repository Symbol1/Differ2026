## 4.6 Variation of Parameters

### When undetermined coefficients stop working

Undetermined coefficients guesses $y_p$ from a finite family  
that contains every derivative of $g$.

For $g = \sec x$ there is no such family:
$$ \begin{aligned}
D \sec x &= \sec(x)\tan x, \\
D^2 \sec x &= \sec(x)^3 + \sec(x)\tan(x)^2, \\
D^3 \sec x &= 5\sec(x)^3 \tan x + \sec(x)\tan(x)^3,
\end{aligned} $$
and each derivative brings a higher power of $\sec x$.  
The same happens for $\tan x$, $\ln x$, and $1/x$.

The method also needs constant coefficients.  
Variation of parameters needs neither:  
it builds $y_p$ from any fundamental set  
$y_1, \ldots, y_n$ of the homogeneous equation.

### First-order equations revisited

For $y' + P(x) y = f(x)$,  
the homogeneous equation has the solution $y_1 = e^{-\int P\,dx}$,  
so $y_c = c_1 y_1$.

Replacing the constant $c_1$ with an unknown function $u(x)$,  
we assume $y_p = u(x) y_1(x)$.  
Since $y_1' + Py_1 = 0$,
$$ y_p' + Py_p = u(y_1' + Py_1) + u' y_1 = u' y_1. $$

So $y_p$ solves the equation exactly when $u' = f/y_1$, and
$$ \boxed{y_p = y_1 \int\frac{f(x)}{y_1(x)}\,dx = e^{-\int P\,dx} \int e^{\int P\,dx} f(x)\,dx.} $$
This is the integrating-factor formula of 2.3.  
The substitution $y = uy_1$ is also the one  
used for reduction of order in 4.2.

### Second-order equations: two varying parameters

Write the equation in standard form,
$$ y'' + P(x) y' + Q(x) y = f(x), $$
and let $y_1$, $y_2$ be a fundamental set  
of the homogeneous equation on $I$.

Replace both constants of $y_c = c_1 y_1 + c_2 y_2$ with functions:
$$ y_p = u_1(x) y_1(x) + u_2(x) y_2(x). $$
The product rule gives
$$ y_p' = u_1 y_1' + u_2 y_2' + (u_1' y_1 + u_2' y_2). $$

We need only one particular solution,  
and one equation cannot determine two unknown functions.  
So we may impose one extra condition.

Choose the one that removes the parenthesis:
$$ \boxed{u_1' y_1 + u_2' y_2 = 0.} $$
Then $y_p' = u_1 y_1' + u_2 y_2'$ and
$$ y_p'' = u_1 y_1'' + u_2 y_2'' + u_1' y_1' + u_2' y_2'. $$
No second derivatives of $u_1$ or $u_2$ appear.

Substituting into the equation and grouping by $u_1$ and $u_2$,
$$ \begin{aligned}
y_p'' + Py_p' + Qy_p &= u_1\underbrace{(y_1'' + Py_1' + Qy_1)}_{=0} \\
&\quad + u_2\underbrace{(y_2'' + Py_2' + Qy_2)}_{=0} \\
&\quad + u_1' y_1' + u_2' y_2'.
\end{aligned} $$

The terms with $u_1$ and $u_2$ cancel because $y_1$ and $y_2$ are solutions,  
as in reduction of order.  
What remains is a second condition:
$$ u_1' y_1' + u_2' y_2' = f(x). $$

### Cramer's rule and the Wronskian

The two conditions form a linear system for $u_1'$ and $u_2'$:
$$ \begin{pmatrix} y_1 & y_2 \\ y_1' & y_2' \end{pmatrix}
\begin{pmatrix} u_1' \\ u_2' \end{pmatrix} = \begin{pmatrix} 0 \\ f(x) \end{pmatrix}. $$
Its determinant is the Wronskian $W = y_1 y_2' - y_1' y_2$,  
which never vanishes on $I$ by 4.1.  
Cramer's rule replaces one column at a time by the right-hand side:
$$ \begin{aligned}
u_1' &= \frac{W_1}{W} = \frac1W\begin{vmatrix} 0 & y_2 \\ f & y_2' \end{vmatrix} = -\frac{y_2 f}{W}, \\
u_2' &= \frac{W_2}{W} = \frac1W\begin{vmatrix} y_1 & 0 \\ y_1' & f \end{vmatrix} = \frac{y_1 f}{W}.
\end{aligned} $$

Integrating,
$$ \boxed{y_p = -y_1 \int\frac{y_2 f}{W}\,dx + y_2 \int\frac{y_1 f}{W}\,dx.} $$
To solve $a_2(x) y'' + a_1(x) y' + a_0(x) y = g(x)$:

1. Find $y_c = c_1 y_1 + c_2 y_2$ and compute $W$.
2. Divide by $a_2(x)$ and read off $f = g/a_2$.
3. Integrate $u_1' = -y_2 f/W$ and $u_2' = y_1 f/W$.
4. Write $y = y_c + u_1 y_1 + u_2 y_2$.

### What the extra condition means

Fix a point $a$ in $I$  
and freeze the parameters at their values there:
$$ Y_a(x) = u_1(a) y_1(x) + u_2(a) y_2(x). $$
Since $u_1(a)$ and $u_2(a)$ are numbers,  
$Y_a$ is a homogeneous solution.  
At $x = a$ it has the same value as $y_p$,  
and by the extra condition also the same slope:
$$ y_p'(a) = u_1(a) y_1'(a) + u_2(a) y_2'(a) = Y_a'(a). $$
So the graph of $Y_a$ touches the graph of $y_p$ at $a$;  
the dashed curves in the picture below are $Y_8$ and $Y_{14}$.

In the state picture of 4.1,  
$\bigl(u_1(x), u_2(x)\bigr)$ are the coordinates of the state $(y_p, y_p')$  
in the basis of the state vectors $(y_1, y_1')$ and $(y_2, y_2')$.

The input enters only through $y''$:  
it changes the slope directly,  
and the value only through the slope.  
This is why the right-hand side of the system is $(0, f)$.

Over a short step $dx$,  
the state follows the frozen solution $Y_a$  
and receives a push $f\,dx$ in the $y'$ direction;  
the equations for $u_1'$ and $u_2'$ record that push  
as a change of the parameters.

Lagrange developed the method for planetary orbits:  
pulled by the other planets,  
a planet always moves on an ellipse whose parameters slowly drift.

#### Picture: an oscillation whose parameters are pushed

Drive $y'' + y = f(x)$ from rest with  
$f = \cos x$ for $0 \le x \le 6\pi$ and $f = 0$ afterwards.  
Here $y_1 = \cos x$, $y_2 = \sin x$, and $W = 1$.

Starting from rest means $y_p(0) = u_1(0) = 0$  
and $y_p'(0) = u_2(0) = 0$, so
$$ \begin{aligned}
u_1(x) &= -\int_0^x \sin(s) f(s)\,ds, \\
u_2(x) &= \int_0^x \cos(s) f(s)\,ds.
\end{aligned} $$

The code integrates these numerically, as it would for any input.  
While the input is on, $y_p = \frac12 x\sin x$: the resonance of 4.4.

```python
import numpy as np
from scipy.integrate import cumulative_trapezoid
push_x = np.linspace(0, 30, 3001)
push_end = float(6*pi)
push_f = np.where(push_x <= push_end, np.cos(push_x), float(0))
push_u1 = cumulative_trapezoid(-np.sin(push_x)*push_f, push_x, initial=0)    # u1' = -y2 f/W
push_u2 = cumulative_trapezoid(np.cos(push_x)*push_f, push_x, initial=0)     # u2' = y1 f/W
push_y = push_u1*np.cos(push_x) + push_u2*np.sin(push_x)
```

```python
push_plot = line(list(zip(push_x, push_f)), color='gray', legend_label='input $f$')
push_plot += line(list(zip(push_x, push_y)), color='black', thickness=2, legend_label='$y_p$')
for a, color in [(8, '#0072B2'), (14, '#009E73')]:
    k = int(np.searchsorted(push_x, a))
    frozen = push_u1[k]*np.cos(push_x) + push_u2[k]*np.sin(push_x)
    push_plot += line(list(zip(push_x, frozen)), color=color, linestyle='--', alpha=0.8,
                      legend_label=f'parameters frozen at $a = {a}$')
    push_plot += point((push_x[k], push_y[k]), color=color, size=60, zorder=5)
push_plot.show(xmin=0, xmax=30, ymin=-11, ymax=15, figsize=[9, 4], frame=True, axes=False,
               axes_labels=['$x$', '$y$'], legend_options={'loc': 'upper left', 'ncol': 2})
```

```python
parameters_plot = line(list(zip(push_x, push_u1)), color='#CC79A7', thickness=2, legend_label='$u_1$')
parameters_plot += line(list(zip(push_x, push_u2)), color='#E69F00', thickness=2, legend_label='$u_2$')
parameters_plot += line([(push_end, -1), (push_end, 10)], color='gray', linestyle=':')
parameters_plot.show(xmin=0, xmax=30, figsize=[9, 3], frame=True, axes=False,
                     axes_labels=['$x$', 'parameters'], legend_options={'loc': 'upper left'})
```

**Read the picture:**  
each dashed curve is a free oscillation with fixed parameters.  
It touches the black curve at its dot  
with the same height and the same slope,  
and then drifts away because the input keeps changing the parameters.  
The input is in phase with the velocity,  
so $u_2 = \frac x2 + \frac14 \sin(2x)$ grows steadily,  
and so does the amplitude.  
At $x = 6\pi$ the input stops and  
the parameters freeze at $u_1 = 0$ and $u_2 = 3\pi$.  
From then on, $y_p = 3\pi\sin x$ is the last frozen oscillation.

### Example 1: $y'' - 4y' + 4y = (x + 1)e^{2x}$

The auxiliary equation $(m - 2)^2 = 0$ gives  
$y_1 = e^{2x}$ and $y_2 = xe^{2x}$, with
$$ W = \begin{vmatrix} e^{2x} & xe^{2x} \\ 2e^{2x} & (2x + 1)e^{2x} \end{vmatrix} = e^{4x}. $$

The equation is already in standard form, so $f = (x + 1)e^{2x}$ and
$$ \begin{aligned}
u_1' &= -\frac{xe^{2x} \cdot (x + 1)e^{2x}}{e^{4x}} = -x^2 - x, \\
u_2' &= \frac{e^{2x} \cdot (x + 1)e^{2x}}{e^{4x}} = x + 1.
\end{aligned} $$

Hence $u_1 = -\frac13 x^3 - \frac12 x^2$  
and $u_2 = \frac12 x^2 + x$, so
$$ \begin{aligned}
y_p &= \Bigl(-\frac13 x^3 - \frac12 x^2\Bigr)e^{2x} + \Bigl(\frac12 x^2 + x\Bigr)xe^{2x} \\
&= \frac16 x^3 e^{2x} + \frac12 x^2 e^{2x}.
\end{aligned} $$
The general solution is
$$ \boxed{y = c_1 e^{2x} + c_2 xe^{2x} + \frac16 x^3 e^{2x} + \frac12 x^2 e^{2x}.} $$

The shift rule of 4.3 gives the same $y_p$ faster:  
$P(D)[e^{2x} h] = e^{2x} D^2 h$,  so $h'' = x + 1$ and $h = \frac16 x^3 + \frac12 x^2$.  
Variation of parameters is the method that always works,  
not always the fastest one.

### Example 2: $4y'' + 36y = \csc(3x)$

First divide by $4$:  
the standard form is $y'' + 9y = \frac14 \csc(3x)$,  so $f = \frac14 \csc(3x)$.  
With $y_1 = \cos(3x)$ and $y_2 = \sin(3x)$,  
the Wronskian is $W = 3$, and
$$ \begin{aligned}
u_1' &= -\frac{\sin(3x) \cdot \frac14 \csc(3x)}{3} = -\frac1{12}, \\
u_2' &= \frac{\cos(3x) \cdot \frac14 \csc(3x)}{3} = \frac{\cos(3x)}{12\sin(3x)}.
\end{aligned} $$
Hence $u_1 = -\frac1{12} x$ and $u_2 = \frac1{36} \ln|\sin(3x)|$.

The general solution is
$$ \boxed{\begin{aligned}
y &= c_1 \cos(3x) + c_2 \sin(3x) \\
&\quad - \frac1{12} x\cos(3x) + \frac1{36} \sin(3x) \ln|\sin(3x)|
\end{aligned}} $$
on every interval where $\sin(3x) \ne 0$, such as $(0, \pi/3)$.  
At the ends of these intervals $\csc(3x)$ blows up.  
The solution stays bounded there,  
since $\sin(3x) \ln|\sin(3x)| \to 0$,  
but its slope contains $\frac1{12} \cos(3x) \ln|\sin(3x)|$ and blows up.  
Forgetting to divide by $4$ would make $y_p$ four times too large.

### Constants of integration

No constants are needed when integrating $u_1'$ and $u_2'$.  
Other antiderivatives $u_1 + a$ and $u_2 + b$ change $y_p$ by $ay_1 + by_2$,  
which is absorbed into $y_c$:
$$ \begin{aligned}
y &= c_1 y_1 + c_2 y_2 + (u_1 + a) y_1 + (u_2 + b) y_2 \\
&= (c_1 + a) y_1 + (c_2 + b) y_2 + u_1 y_1 + u_2 y_2.
\end{aligned} $$
For the same reason,  
any term of $y_p$ that is a multiple of $y_1$ or $y_2$ may be dropped.

### Example 3: $y'' - y = 1/x$

Here $y_1 = e^x$, $y_2 = e^{-x}$, and $W = -2$, so
$$ \begin{aligned}
u_1' &= \frac{e^{-x}}{2x}, \\
u_2' &= -\frac{e^x}{2x}.
\end{aligned} $$
These have no elementary antiderivatives.

As in 2.3, keep them as integral-defined functions,  
with a base point $x_0$ on the same side of $0$ as $x$:
$$ y_p = \frac12 e^x \int_{x_0}^x \frac{e^{-t}}{t}\,dt - \frac12 e^{-x} \int_{x_0}^x \frac{e^t}{t}\,dt. $$

Moving $e^{\pm x}$ inside the integrals combines them into one:
$$ y_p = \int_{x_0}^x \frac{e^{x-t} - e^{-(x-t)}}{2t}\,dt = \int_{x_0}^x \frac{\sinh(x - t)}{t}\,dt. $$
The general solution on $(0, \infty)$ or on $(-\infty, 0)$ is
$$ \boxed{y = c_1 e^x + c_2 e^{-x} + \int_{x_0}^x \frac{\sinh(x - t)}{t}\,dt.} $$

### The particular solution as one integral

The same combination works in general.  
With definite integrals from $x_0$,
$$ \begin{aligned}
y_p(x) &= \int_{x_0}^x \frac{y_1(t) y_2(x) - y_1(x) y_2(t)}{W(t)} f(t)\,dt \\
&= \int_{x_0}^x G(x, t) f(t)\,dt.
\end{aligned} $$
For fixed $t$, the kernel $G(x, t)$  
is a homogeneous solution in $x$ with
$$ G(t, t) = 0, \qquad \frac{\partial G}{\partial x}(t, t) = 1. $$
This $y_p$ satisfies $y_p(x_0) = y_p'(x_0) = 0$:  
it is the zero-state response of 4.1.

Read the formula as superposition:  
split the input into short pieces.  
The piece on $[t, t + dt]$ alone kicks the slope by $f(t)\,dt$  
and leaves the value unchanged.

Suppose the piece at $t = 0$ is the only kick,  
like giving a swing a single push at midnight.  
After the push the swing moves freely:  
the solution is the homogeneous solution  
with value $0$ and slope $f(0)\,dt$ at $x = 0$, which is $G(x, 0) f(0)\,dt$.  
A kick at any other time $t$ works the same way  
and contributes $G(x, t) f(t)\,dt$.

By linearity, the particular solution is  
the sum of the effects of all the tiny kicks.  
The kernel $G(x, t)$ tells how a kick at time $t$ shows up at a later $x$;  
it is called the **Green's function** of the initial value problem.

For constant coefficients, $G$ depends only on $x - t$:

| Equation | $G(x, t)$ |
|---|---|
| $y'' + k^2 y = f$ | $\sin\bigl(k(x - t)\bigr)/k$ |
| $y'' - k^2 y = f$ | $\sinh\bigl(k(x - t)\bigr)/k$ |
| $y'' - 2my' + m^2 y = f$ | $(x - t)e^{m(x-t)}$ |

With $m = -1/\tau$, the last kernel is, up to the factor $1/\tau^2$,  
the weight of two EMA stages in the bonus section of 1021.

### Higher-order equations

For $y^{(n)} + p_{n-1}(x) y^{(n-1)} + \cdots + p_0(x) y = f(x)$  
with a fundamental set $y_1, \ldots, y_n$, try
$$ y_p = u_1 y_1 + u_2 y_2 + \cdots + u_n y_n. $$
Impose $n - 1$ conditions so that  
$y_p', \ldots, y_p^{(n-1)}$ contain no derivatives of the $u_k$,  
exactly as for $n = 2$.

Together with the equation itself, the conditions read
$$ \begin{pmatrix} y_1 & y_2 & \cdots & y_n \\ y_1' & y_2' & \cdots & y_n' \\
\vdots & \vdots & & \vdots \\ y_1^{(n-1)} & y_2^{(n-1)} & \cdots & y_n^{(n-1)} \end{pmatrix}
\begin{pmatrix} u_1' \\ u_2' \\ \vdots \\ u_n' \end{pmatrix} = \begin{pmatrix} 0 \\ \vdots \\ 0 \\ f(x) \end{pmatrix}. $$

The matrix is again the Wronskian matrix, so Cramer's rule gives
$$ \boxed{u_k' = \frac{W_k}{W}, \qquad k = 1, \ldots, n,} $$
where $W_k$ is the determinant obtained from $W$  
by replacing column $k$ with $(0, \ldots, 0, f)$.  
Expanding $W_k$ along that column,  
only the entry $f$ in the last row contributes.

### Example 4: $y''' + y' = \tan x$

The auxiliary equation $m(m^2 + 1) = 0$  
gives $y_1 = 1$, $y_2 = \cos x$, $y_3 = \sin x$, and
$$ W = \begin{vmatrix} 1 & \cos x & \sin x \\ 0 & -\sin x & \cos x \\ 0 & -\cos x & -\sin x \end{vmatrix} = 1. $$
Replacing one column at a time by $(0, 0, \tan x)$,
$$ \begin{aligned}
W_1 &= \tan x, \\
W_2 &= -\cos(x)\tan x = -\sin x, \\
W_3 &= -\sin(x)\tan x.
\end{aligned} $$

On $(-\pi/2, \pi/2)$, where $\cos x > 0$, integration gives
$$ \begin{aligned}
u_1 &= \ln(\sec x), \\
u_2 &= \cos x, \\
u_3 &= \sin x - \ln(\sec x + \tan x),
\end{aligned} $$
using $\sin(x)\tan x = \sec x - \cos x$.

Then $u_1 y_1 + u_2 y_2 + u_3 y_3$  
$= \ln(\sec x) + 1 - \sin(x) \ln(\sec x + \tan x)$.  
Drop the constant $1$, which belongs to $y_c$:
$$ \boxed{\begin{aligned}
y &= c_1 + c_2 \cos x + c_3 \sin x \\
&\quad + \ln(\sec x) - \sin(x) \ln(\sec x + \tan x).
\end{aligned}} $$

### Example 5: combining the two methods

Solve $y'' - 2y' + y = 4x^2 - 3 + x^{-1} e^x$,  
where $y_c = c_1 e^x + c_2 xe^x$.  
By superposition (4.1), treat the two parts of the input separately  
and use the easier method for each.

- For $4x^2 - 3$, undetermined coefficients  
  with $Ax^2 + Bx + C$ gives $A = 4$, $B = 16$, $C = 21$.
- For $x^{-1} e^x$, variation of parameters  
  with $W = e^{2x}$ gives $u_1' = -1$ and $u_2' = 1/x$,  
  so $u_1 = -x$ and $u_2 = \ln|x|$.

The second part gives $-xe^x + xe^x \ln|x|$,  
and $-xe^x$ belongs to $y_c$.  
Therefore
$$ \boxed{y = c_1 e^x + c_2 xe^x + 4x^2 + 16x + 21 + xe^x \ln|x|} $$
on $(0, \infty)$ or on $(-\infty, 0)$.  
Undetermined coefficients alone fails because of $x^{-1} e^x$.  
Variation of parameters alone works,  
but the polynomial part would need several integrations by parts.

### Choosing a method

| Requirement | Undetermined coefficients (4.4, 4.5) | Variation of parameters (4.6) |
|---|---|---|
| Coefficients | constant | any, once $y_c$ is known |
| Input | polynomials, exponentials, sines, cosines, and their products | any continuous $f$ |
| Work | a linear system for numbers | integrals, possibly nonelementary |
| Standard form | not needed | needed: $f = g/a_n$ |

## 4.7 Cauchy–Euler Equation

### Equations whose powers match the derivatives

A **Cauchy–Euler equation** has the form
$$ a_n x^n y^{(n)} + a_{n-1} x^{n-1} y^{(n-1)} + \cdots + a_1 xy' + a_0 y = g(x), $$
with constants $a_0, \ldots, a_n$:  
the power of $x$ in each coefficient  
equals the order of the derivative.  
Several equations of 4.1 and 4.2 were of this type, such as
$$ \begin{aligned}
x^2 y'' - 3xy' + 4y &= 0, \\
x^2 y'' - 4xy' + 6y &= 0, \\
x^3 y''' - 2xy' + 4y &= 0.
\end{aligned} $$
Their solutions $x^2$, $x^2 \ln x$, $x^3$ were given there;  
this section explains where they come from.

The leading coefficient $a_n x^n$ vanishes at $x = 0$,  
the only singular point.  
We first solve on $(0, \infty)$ and treat $(-\infty, 0)$ at the end.  
The rescaling $X = cx$ with a constant $c > 0$  
leaves every term unchanged:
$$ x^k \frac{d^k y}{dx^k} = X^k \frac{d^k y}{dX^k}. $$
The equation has no preferred length scale,  
so it is also called **equidimensional**.  
Constant-coefficient equations have no preferred origin  
and are solved by exponentials;  
scale-free equations are solved by powers.

### Powers are eigenfunctions of $xD$

For $y = x^m$,
$$ x^k D^k x^m = m(m - 1)\cdots(m - k + 1) x^m, $$
so every term of the equation is a polynomial in $m$ times $x^m$.

For the second-order equation,
$$ ax^2 y'' + bxy' + cy = \bigl(am(m - 1) + bm + c\bigr)x^m. $$
Hence $x^m$ solves the homogeneous equation  
exactly when $m$ solves the **auxiliary equation**
$$ \boxed{\begin{aligned}
am(m - 1) + bm + c &= 0, \\
\text{that is,}\quad am^2 + (b - a)m + c &= 0.
\end{aligned}} $$
The coefficient of $m$ is $b - a$, not $b$.

In operator language, write $\theta = xD$.  
Then $\theta x^m = mx^m$:  
the power $x^m$ is an eigenfunction of $\theta$,  
as $e^{mx}$ is an eigenfunction of $D$ in 4.3.

Below we show that $x^k D^k = \theta(\theta - 1)\cdots(\theta - k + 1)$,  
so a Cauchy–Euler equation reads $P(\theta) y = g$ with
$$ \begin{aligned}
P(m) &= \sum_{k=0}^n a_k m(m - 1)\cdots(m - k + 1), \\
P(\theta) x^m &= P(m) x^m.
\end{aligned} $$
This is the analogue of $P(D) e^{mx} = P(m) e^{mx}$,  
and the three cases of 4.3 reappear.

### Case I: distinct real roots

If the auxiliary equation has real roots $m_1 \ne m_2$, then
$$ \boxed{y = c_1 x^{m_1} + c_2 x^{m_2}.} $$
The two powers are independent on $(0, \infty)$,  
since $W(x^{m_1}, x^{m_2}) = (m_2 - m_1)x^{m_1+m_2-1} \ne 0$.

### Example 1: $x^2 y'' - 2xy' - 4y = 0$

Rather than memorize the auxiliary equation,  
substitute $y = x^m$ directly:
$$ \begin{aligned}
x^2 y'' - 2xy' - 4y &= x^2 \cdot m(m - 1)x^{m-2} \\
&\quad - 2x \cdot mx^{m-1} - 4x^m \\
&= (m^2 - 3m - 4)x^m \\
&= (m + 1)(m - 4)x^m.
\end{aligned} $$
So $m_1 = -1$, $m_2 = 4$, and
$$ \boxed{y = c_1 x^{-1} + c_2 x^4.} $$

### Case II: a repeated real root

If $(b - a)^2 = 4ac$, there is a single root $m_1 = -(b - a)/(2a)$,  
so $2m_1 = 1 - b/a$.  
For reduction of order,  
the standard form $y'' + \frac{b}{ax} y' + \frac{c}{ax^2} y = 0$  
gives $e^{-\int P\,dx} = x^{-b/a}$, and
$$ \begin{aligned}
y_2 &= x^{m_1} \int\frac{x^{-b/a}}{x^{2m_1}}\,dx = x^{m_1} \int\frac{x^{-b/a}}{x^{1-b/a}}\,dx \\
&= x^{m_1} \int\frac{dx}{x} = x^{m_1} \ln x.
\end{aligned} $$
Therefore
$$ \boxed{y = c_1 x^{m_1} + c_2 x^{m_1} \ln x.} $$

For $x^2 y'' - 3xy' + 4y = 0$,  
the auxiliary equation is $m^2 - 4m + 4 = (m - 2)^2 = 0$,  
which explains $x^2$ and $x^2 \ln x$ in Example 2 of 4.2.  
The factor $\ln x$ plays the role of the factor $x$  
for repeated roots in 4.3.  
A root $m_1$ of multiplicity $k$ gives the $k$ solutions
$$ x^{m_1}, \qquad x^{m_1} \ln x, \qquad \ldots, \qquad x^{m_1} \ln(x)^{k-1}. $$

### Example 2: $4x^2 y'' + 8xy' + y = 0$

The substitution $y = x^m$ gives
$$ 4m(m - 1) + 8m + 1 = 4m^2 + 4m + 1 = (2m + 1)^2 = 0, $$
so $m_1 = -\frac12$ is a double root and
$$ \boxed{y = c_1 x^{-1/2} + c_2 x^{-1/2} \ln x.} $$

### Case III: complex conjugate roots

If the roots are $\alpha \pm i\beta$ with $\beta > 0$,  
write $x = e^{\ln x}$ and use Euler's formula:
$$ x^{\alpha+i\beta} = x^\alpha e^{i\beta\ln x} = x^\alpha \cos(\beta\ln x) + ix^\alpha \sin(\beta\ln x). $$
The coefficients of the equation are real,  
so the real and imaginary parts are real solutions:
$$ \boxed{y = x^\alpha\bigl(c_1 \cos(\beta\ln x) + c_2 \sin(\beta\ln x)\bigr).} $$
Their Wronskian is $\beta x^{2\alpha-1} \ne 0$ on $(0, \infty)$.  
The oscillation is periodic in $\ln x$, not in $x$.

### Example 3: an initial value problem

Solve
$$ 4x^2 y'' + 17y = 0, \qquad y(1) = -1, \qquad y'(1) = -\frac12. $$
The $y'$ term is missing,  
but $y = x^m$ still gives $4m(m - 1) + 17 = 4m^2 - 4m + 17 = 0$.  
So $m = \frac12 \pm 2i$ and
$$ y = x^{1/2}\bigl(c_1 \cos(2\ln x) + c_2 \sin(2\ln x)\bigr). $$
Since $\ln1 = 0$, the conditions give $y(1) = c_1 = -1$  
and $y'(1) = \frac12 c_1 + 2c_2 = -\frac12$, so $c_2 = 0$:
$$ \boxed{y = -x^{1/2} \cos(2\ln x).} $$

#### Picture: a clock that ticks in $\ln x$

The zeros of $y = -x^{1/2} \cos(2\ln x)$  
satisfy $2\ln x = \frac\pi2 + k\pi$,  
that is, $x_k = e^{\pi/4+k\pi/2}$ for every integer $k$.  
Consecutive zeros have the constant ratio $e^{\pi/2} \approx 4.81$  
instead of a constant spacing.  
Plot the solution against a linear $x$-axis,  
and then its ratio to the envelope $x^{1/2}$  
against a logarithmic one.

```python
clock_x = np.logspace(-3, 2, 3001)
clock_envelope = np.sqrt(clock_x)
clock_zeros = [float(exp(pi/4 + k*pi/2)) for k in range(-4, 3)]
def clock_plot(values, envelope, label):
    result = line(list(zip(clock_x, values)), color='#0072B2', thickness=1.5, legend_label=label)
    for sign in [1, -1]:
        result += line(list(zip(clock_x, sign*envelope)), color='gray', linestyle='--')
    return result + points([(z, 0) for z in clock_zeros], color='#D55E00', size=30, zorder=5,
                           legend_label='zeros')
```

```python
clock_cos = -np.cos(2*np.log(clock_x))
clock_plot(clock_envelope*clock_cos, clock_envelope, r'$y = -x^{1/2}\cos(2\ln x)$').show(
    xmin=0, xmax=60, ymin=-8.5, ymax=8.5, figsize=[9, 3.5], frame=True, axes=False,
    axes_labels=['$x$', '$y$'], legend_options={'loc': 'upper left'})
clock_plot(clock_cos, np.ones_like(clock_x), r'$y/x^{1/2} = -\cos(2\ln x)$').show(
    scale='semilogx', xmin=1e-3, xmax=60, ymin=-1.3, ymax=1.9, figsize=[9, 3.5], frame=True, axes=False,
    axes_labels=['$x$ (logarithmic)', '$y/x^{1/2}$'], legend_options={'loc': 'upper left', 'ncol': 2})
```

**Read the picture:**  
on the linear axis the zeros crowd toward $x = 0$,  
with infinitely many of them in $(0, 1)$,  
while the gaps grow by the factor $4.81$ to the right.  
After dividing by the envelope,  
the logarithmic axis shows a pure cosine with evenly spaced zeros,  
$\pi/2$ apart in $t = \ln x$.  
In the variable $t$ the solution is $-e^{t/2} \cos(2t)$,  
a solution of a constant-coefficient equation.  
This suggests the substitution below.

### Example 4: a third-order equation

For $x^3 y''' + 5x^2 y'' + 7xy' + 8y = 0$,  
the substitution $y = x^m$ gives the auxiliary polynomial
$$ \begin{aligned}
P(m) &= m(m - 1)(m - 2) + 5m(m - 1) + 7m + 8 \\
&= m^3 + 2m^2 + 4m + 8 = (m + 2)(m^2 + 4).
\end{aligned} $$
The roots are $-2$ and $\pm2i$, so
$$ \boxed{y = c_1 x^{-2} + c_2 \cos(2\ln x) + c_3 \sin(2\ln x).} $$

### Reduction to constant coefficients

On $(0, \infty)$ substitute $x = e^t$, that is, $t = \ln x$.  
The chain rule  
gives $\frac{dy}{dt} = \frac{dx}{dt}\frac{dy}{dx} = x\frac{dy}{dx}$, so
$$ \boxed{xD_x = D_t.} $$
The operator $\theta = xD$ is differentiation in $t$.  
Moreover, $D_t(x^k D_x^k y) = xD_x(x^k D_x^k y)$  
$= kx^k D_x^k y + x^{k+1} D_x^{k+1} y$, so
$$ x^{k+1} D_x^{k+1} = (D_t - k) x^k D_x^k. $$
Starting from $xD_x = D_t$ and repeating,  
$x^k D_x^k = D_t(D_t - 1)\cdots(D_t - k + 1)$.

In particular $x^2 D_x^2 = D_t^2 - D_t$,  
and the second-order Cauchy–Euler equation becomes
$$ \begin{aligned}
ax^2 y'' + bxy' + cy &= g(x) \\
\Longrightarrow\quad a\frac{d^2y}{dt^2} + (b - a)\frac{dy}{dt} + cy &= g(e^t).
\end{aligned} $$
The new equation has constant coefficients,  
and its auxiliary polynomial is the same $am^2 + (b - a)m + c$.  
Back in $x$, every formula of 4.3 translates through $e^{mt} = x^m$.

| Constant coefficients in $t$ | Cauchy–Euler in $x = e^t$ |
|---|---|
| $e^{mt}$ | $x^m$ |
| $te^{mt}$ | $x^m \ln x$ |
| $e^{\alpha t} \cos(\beta t)$, $e^{\alpha t} \sin(\beta t)$ | $x^\alpha \cos(\beta\ln x)$, $x^\alpha \sin(\beta\ln x)$ |
| translation $t \mapsto t + s$ | scaling $x \mapsto e^s x$ |
| $D_t$ | $\theta = xD_x$ |

The three cases of this section  
are the three cases of 4.3 in disguise.

### Example 5: $x^2 y'' - xy' + y = \ln x$

With $t = \ln x$, the equation becomes
$$ \frac{d^2y}{dt^2} - 2\frac{dy}{dt} + y = t, $$
whose auxiliary equation is $(m - 1)^2 = 0$,  
so $y_c = c_1 e^t + c_2 te^t$.  
The trial $y_p = A + Bt$ gives $-2B + A + Bt = t$,  
so $B = 1$ and $A = 2$.  
Returning to $x$,
$$ \boxed{y = c_1 x + c_2 x\ln x + 2 + \ln x.} $$

### Inputs that are powers of $x$

An input $x^r = e^{rt}$ is an exponential in $t$,  
so 4.4 applies without changing variables.  
Since $P(\theta) x^r = P(r) x^r$,  
a particular solution of $P(\theta) y = x^r$  
is $x^r/P(r)$ when $P(r) \ne 0$.  
If $r$ is a root of $P$ of multiplicity $k$,  
the formula of 4.4 with $t^k = \ln(x)^k$ gives
$$ \boxed{P(\theta) y = x^r \quad\Longrightarrow\quad y_p = \frac{x^r \ln(x)^k}{P^{(k)}(r)}.} $$
Here $P(\theta) y$ is the equation in its Cauchy–Euler form,  
before dividing by the leading coefficient.

### Example 6: $x^2 y'' - 3xy' + 3y = x^3 + 6$

Here $P(m) = m(m - 1) - 3m + 3 = (m - 1)(m - 3)$,  
so $y_c = c_1 x + c_2 x^3$.

- For $6 = 6x^0$: $P(0) = 3$, so $y_{p_1} = 2$.
- For $x^3$: $3$ is a simple root,  
  and $P'(m) = 2m - 4$ gives $P'(3) = 2$,  
  so $y_{p_2} = \frac12 x^3 \ln x$.

Therefore
$$ \boxed{y = c_1 x + c_2 x^3 + \frac12 x^3 \ln x + 2.} $$

In the same way, $x^2 y'' - 2xy' + 2y = 6$  
from 4.1 has $P(m) = (m - 1)(m - 2)$ and $y_p = 6/P(0) = 3$.  
This explains  
why all its solutions $c_1 x + c_2 x^2 + 3$ pass through $(0, 3)$.

### Example 7: variation of parameters

For $x^2 y'' - 3xy' + 3y = 2x^4 e^x$, the input is not a power of $x$,  
and undetermined coefficients does not apply.  
From Example 6, $y_1 = x$ and $y_2 = x^3$, with $W = 2x^3$.  
Divide by $x^2$ first:  
the standard form has $f = 2x^2 e^x$, not $2x^4 e^x$.  
Then
$$ \begin{aligned}
u_1' &= -\frac{x^3 \cdot 2x^2 e^x}{2x^3} = -x^2 e^x, \\
u_2' &= \frac{x \cdot 2x^2 e^x}{2x^3} = e^x.
\end{aligned} $$

Integrating by parts twice,  
$u_1 = -(x^2 - 2x + 2)e^x$, and $u_2 = e^x$, so
$$ y_p = -(x^3 - 2x^2 + 2x)e^x + x^3 e^x = 2x^2 e^x - 2xe^x. $$
The general solution on $(0, \infty)$ is
$$ \boxed{y = c_1 x + c_2 x^3 + 2x^2 e^x - 2xe^x.} $$

### Solutions for $x < 0$

On $(-\infty, 0)$ substitute $s = -x > 0$.  
Then $\frac{d}{dx} = -\frac{d}{ds}$, so
$$ x^k \frac{d^k y}{dx^k} = (-s)^k (-1)^k \frac{d^k y}{ds^k} = s^k \frac{d^k y}{ds^k}, $$
and a homogeneous Cauchy–Euler equation  
keeps exactly the same form in $s$.  
Hence the solutions on $(-\infty, 0)$ are those  
on $(0, \infty)$ with $x$ replaced by $|x|$, for example
$$ \begin{aligned}
y &= c_1 |x|^{m_1} + c_2 |x|^{m_2} \qquad\text{or} \\
y &= |x|^\alpha\bigl(c_1 \cos(\beta\ln|x|) + c_2 \sin(\beta\ln|x|)\bigr).
\end{aligned} $$
Written with $|x|$, these formulas hold on either side of $0$.

### Back to $x^3$ and $|x|^3$

In 4.1, $x^3$ and $|x|^3$ both solve $x^2 y'' - 4xy' + 6y = 0$,  
and their Wronskian vanishes everywhere,  
yet they are independent on $(-\infty, \infty)$.  
Now we can find all solutions.  
The auxiliary equation $m^2 - 5m + 6 = (m - 2)(m - 3) = 0$ gives
$$ y = \begin{cases} c_1 x^2 + c_2 x^3, & x > 0, \\ d_1 x^2 + d_2 x^3, & x < 0. \end{cases} $$

Both pieces tend to $0$ at $x = 0$  
together with their first derivatives,  
their second derivatives tend to $2c_1$ and $2d_1$,  
and the equation at $x = 0$ only requires $6y(0) = 0$.  
So $y$ is a solution on all of $(-\infty, \infty)$  
exactly when $c_1 = d_1$:  
the solutions form a three-dimensional space,  
spanned by $x^2$, $x^3$, $|x|^3$.  
The count $\dim = n$ of 4.1 fails across the singular point,  
and initial data at $x_0 = 0$ cannot select a solution,  
since every solution has $y(0) = y'(0) = 0$.

### A shifted center

The equation
$$ a(x - x_0)^2 y'' + b(x - x_0) y' + cy = 0 $$
is a Cauchy–Euler equation in the variable $s = x - x_0$,  
with singular point $x_0$.  
Try $y = (x - x_0)^m$;  
the auxiliary equation is again $am^2 + (b - a)m + c = 0$.  
For example, $(x + 2)^2 y'' + (x + 2) y' + y = 0$  
gives $m^2 + 1 = 0$ and
$$ y = c_1 \cos\bigl(\ln(x + 2)\bigr) + c_2 \sin\bigl(\ln(x + 2)\bigr), \qquad x > -2. $$

### Summary

| Task | Tool |
|---|---|
| $y_p$ of $y'' + Py' + Qy = f$ for any continuous $f$ | $u_1' = -y_2 f/W$, $u_2' = y_1 f/W$, $y_p = u_1 y_1 + u_2 y_2$ |
| $y_p$ starting from rest at $x_0$ | $y_p = \int_{x_0}^x G(x, t) f(t)\,dt$ |
| $y_p$ for order $n$ | $u_k' = W_k/W$ by Cramer's rule |
| Solve a Cauchy–Euler equation | $y = x^m$ with $am(m - 1) + bm + c = 0$; $\ln x$ for repeated roots; $x^\alpha \cos(\beta\ln x)$, $x^\alpha \sin(\beta\ln x)$ for complex pairs |
| Cauchy–Euler with input $x^r$ or $\ln x$ | $x = e^t$, or $y_p = x^r \ln(x)^k/P^{(k)}(r)$ |
| Cauchy–Euler with other inputs | variation of parameters, after dividing by $a_n x^n$ |
| Solutions for $x < 0$ | replace $x$ by $\lvert x\rvert$ |

### Practice habits for variation of parameters and Cauchy–Euler equations

- Divide by the leading coefficient before reading off $f$;  
  for a Cauchy–Euler equation, divide by $ax^2$.
- Simplify $W$ before dividing by it.
- Integrate $u_1'$ and $u_2'$ without constants,  
  and drop terms of $y_p$ that belong to $y_c$.
- Use undetermined coefficients  
  for the parts of the input it can handle.
- For a Cauchy–Euler equation, substitute $y = x^m$ term by term;  
  the coefficient of $m$ is $b - a$.
- State the interval: avoid $x = 0$  
  and the points where $f$ is discontinuous.

### Exam vocabulary

| Phrase | 中文 | What to write |
|---|---|---|
| **variation of parameters** | 參數變異法 | $y_p = u_1 y_1 + u_2 y_2$ with $u_1' y_1 + u_2' y_2 = 0$ |
| **Cramer's rule** | 克拉瑪法則 | $u_k' = W_k/W$ |
| **standard form** | 標準形式 | leading coefficient $1$; read off $f$ there |
| **Cauchy–Euler equation**, **equidimensional equation** | 柯西–歐拉方程、等維方程 | only terms $a_k x^k y^{(k)}$ |
| **Green's function** | 格林函數 | the kernel $G(x, t)$ in $y_p = \int G(x, t) f(t)\,dt$ |

## Checking Worked Solutions with Sage

Sage checks residuals,  
solves the linear systems of variation of parameters,  
factors auxiliary polynomials, and solves initial value problems.  
As before, a symbolic check confirms the algebra;  
dividing by the leading coefficient  
and choosing the interval are still our job.

```python
from IPython.display import display, Math
x = SR.var('x')
c1, c2, c3 = SR.var('c1 c2 c3')
def apply_L(coefficients, y):
    # coefficients = [a_0, a_1, ..., a_n] for a_n y^(n) + ... + a_1 y' + a_0 y
    return sum(a*diff(SR(y), x, k) for k, a in enumerate(coefficients))
```

```python
residual_checks = [
    ('4.6 Ex 1', [4, -4, 1], x**3*exp(2*x)/6 + x**2*exp(2*x)/2, (x + 1)*exp(2*x)),
    ('4.6 Ex 2', [36, 0, 4], -x*cos(3*x)/12 + sin(3*x)*log(sin(3*x))/36, csc(3*x)),
    ('4.6 Ex 3', [-1, 0, 1], (exp(x)*Ei(-x) - exp(-x)*Ei(x))/2, 1/x),
    ('4.6 Ex 4', [0, 1, 0, 1], log(sec(x)) - sin(x)*log(sec(x) + tan(x)), tan(x)),
    ('4.6 Ex 5', [1, -2, 1], 4*x**2 + 16*x + 21 + x*exp(x)*log(x), 4*x**2 - 3 + exp(x)/x),
    ('4.6 picture', [1, 0, 1], x*sin(x)/2, cos(x)),
]
```

```python
residual_checks += [
    ('4.7 Ex 1', [-4, -2*x, x**2], c1/x + c2*x**4, 0),
    ('4.7 Ex 2', [1, 8*x, 4*x**2], (c1 + c2*log(x))/sqrt(x), 0),
    ('4.7 Ex 3', [17, 0, 4*x**2], -sqrt(x)*cos(2*log(x)), 0),
    ('4.7 Ex 4', [8, 7*x, 5*x**2, x**3], c1/x**2 + c2*cos(2*log(x)) + c3*sin(2*log(x)), 0),
    ('4.7 Ex 5', [1, -x, x**2], c1*x + c2*x*log(x) + 2 + log(x), log(x)),
    ('4.7 Ex 6', [3, -3*x, x**2], x**3*log(x)/2 + 2, x**3 + 6),
    ('4.7 Ex 7', [3, -3*x, x**2], 2*x**2*exp(x) - 2*x*exp(x), 2*x**4*exp(x)),
]
```

```python
assume(x > 0)
for name, coefficients, y, g in residual_checks:
    assert (apply_L(coefficients, y) - g).simplify_full().simplify_trig() == 0, name
forget()
print('Residuals vanish:', ', '.join(name for name, *_ in residual_checks))
```

```python
def parameter_derivatives(ys, f):
    # W and u_1', ..., u_n' from the Wronskian system; f is the right side of the standard form
    n = len(ys)
    M = matrix(SR, [[diff(y, x, i) for y in ys] for i in range(n)])
    derivatives = M.solve_right(vector(SR, [0]*(n - 1) + [f]))
    return M.det().simplify_trig(expand=False), [d.simplify_trig(expand=False) for d in derivatives]
vp_examples = [('4.6 Ex 1', [exp(2*x), x*exp(2*x)], (x + 1)*exp(2*x)),
               ('4.6 Ex 2', [cos(3*x), sin(3*x)], csc(3*x)/4),
               ('4.6 Ex 3', [exp(x), exp(-x)], 1/x),
               ('4.6 Ex 4', [1, cos(x), sin(x)], tan(x)),
               ('4.7 Ex 7', [x, x**3], 2*x**2*exp(x))]
```

```python
for name, ys, f in vp_examples:
    W, derivatives = parameter_derivatives(ys, f)
    display(Math(r'\text{' + name + r'}: \quad W = ' + latex(W) + r', \quad '
                 + r', \quad '.join(f"u_{k+1}' = " + latex(d) for k, d in enumerate(derivatives))))
```

These agree with Examples 1–4 of 4.6 and Example 7 of 4.7.  
The antiderivatives are left to us on purpose.  
For $\cot(3x)$ and $\sin(x)\tan x$,  
Sage writes them with logarithms of $\cos x - 1$ and $\sin x - 1$,  
which are negative on our intervals.  
The residual checks above  
confirm the real antiderivatives found by hand.

```python
R = PolynomialRing(ZZ, 'm'); m = R.gen()
def euler_polynomial(coefficients):
    # the term a_k x^k y^(k) contributes a_k m(m - 1)...(m - k + 1)
    return sum(a*falling_factorial(m, k) for k, a in enumerate(coefficients))
for coefficients in [[-4, -2, 1], [1, 8, 4], [17, 0, 4], [8, 7, 5, 1], [3, -3, 1], [6, -4, 1]]:
    P = euler_polynomial(coefficients)
    display(Math(latex(P) + ' = ' + latex(P.factor())))
```

These are the auxiliary polynomials of Examples 1–4 and 6 of 4.7,  
and of $x^2 y'' - 4xy' + 6y = 0$.
